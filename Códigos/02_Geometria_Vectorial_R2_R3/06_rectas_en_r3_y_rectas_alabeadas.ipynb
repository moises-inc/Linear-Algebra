## 1. Fundamentos Teóricos

### 1.1. Ecuaciones de la Recta en $\mathbb{R}^3$
Una recta $L$ en el espacio tridimensional queda unívocamente determinada por un punto base $P_0(x_0, y_0, z_0)$ y un vector director no nulo $\mathbf{d} = (d_x, d_y, d_z)$:

1. **Ecuación Vectorial:**
   $$\mathbf{r}(t) = P_0 + t\,\mathbf{d} = (x_0, y_0, z_0) + t(d_x, d_y, d_z), \quad t \in \mathbb{R}$$

2. **Ecuaciones Paramétricas Escalares:**
   $$\begin{cases} x(t) = x_0 + t\,d_x \\ y(t) = y_0 + t\,d_y \\ z(t) = z_0 + t\,d_z \end{cases}$$

3. **Ecuaciones Simétricas** (suponiendo componentes directores no nulos):
   $$\frac{x - x_0}{d_x} = \frac{y - y_0}{d_y} = \frac{z - z_0}{d_z}$$

---

### 1.2. Clasificación de Posiciones Relativas de dos Rectas
Dadas $L_1: \mathbf{r}_1(t) = P_1 + t\,\mathbf{d}_1$ y $L_2: \mathbf{r}_2(s) = P_2 + s\,\mathbf{d}_2$:

- Si $\mathbf{d}_1 \times \mathbf{d}_2 = \mathbf{0}$ (directores paralelos):
  - Si $(P_2 - P_1) \times \mathbf{d}_1 = \mathbf{0} \implies$ **Coincidentes** (misma recta).
  - Si $(P_2 - P_1) \times \mathbf{d}_1 \neq \mathbf{0} \implies$ **Paralelas no coincidentes**.
- Si $\mathbf{d}_1 \times \mathbf{d}_2 \neq \mathbf{0}$ (directores no paralelos):
  - Si $(P_2 - P_1) \cdot (\mathbf{d}_1 \times \mathbf{d}_2) = 0 \implies$ **Secantes** (intersecan en un punto único).
  - Si $(P_2 - P_1) \cdot (\mathbf{d}_1 \times \mathbf{d}_2) \neq 0 \implies$ **Alabeadas / Skew** (no se cortan ni son paralelas).

---

### 1.3. Distancia Mínima entre Rectas Alabeadas
Dado que las rectas no se intersecan ni son paralelas, residen en dos planos paralelos $\Pi_1$ y $\Pi_2$ cuya normal común es $\mathbf{n} = \mathbf{d}_1 \times \mathbf{d}_2$. La distancia entre ambas es la proyección del vector $\vec{P_1 P_2}$ sobre dicha dirección normal:
$$d(L_1, L_2) = \frac{|(P_2 - P_1) \cdot (\mathbf{d}_1 \times \mathbf{d}_2)|}{\|\mathbf{d}_1 \times \mathbf{d}_2\|}$$

> [!NOTE] Invarianza de la Distancia Mínima y Planos Paralelos
> La distancia mínima $d(L_1, L_2)$ entre dos rectas alabeadas es estrictamente positiva ($d > 0$). Coincide exactamente con la distancia perpendicular entre los dos planos paralelos $\Pi_1$ y $\Pi_2$ que las contienen. El vector $\vec{Q_1 Q_2}$ es simultáneamente perpendicular a $\mathbf{d}_1$ y $\mathbf{d}_2$.

---

### 1.4. Determinación del Segmento Perpendicular Común (Sistema de Gram)
Buscamos los parámetros óptimos $t^*, s^*$ tales que el vector de enlace $\mathbf{w} = L_2(s) - L_1(t) = (P_2 - P_1) + s\mathbf{d}_2 - t\mathbf{d}_1$ sea ortogonal a ambos vectores directores:
$$\begin{cases} \mathbf{w} \cdot \mathbf{d}_1 = 0 \\ \mathbf{w} \cdot \mathbf{d}_2 = 0 \end{cases} \implies \begin{pmatrix} \|\mathbf{d}_1\|^2 & -(\mathbf{d}_1 \cdot \mathbf{d}_2) \\ \mathbf{d}_1 \cdot \mathbf{d}_2 & -\|\mathbf{d}_2\|^2 \end{pmatrix} \begin{pmatrix} t \\ s \end{pmatrix} = \begin{pmatrix} (P_2 - P_1) \cdot \mathbf{d}_1 \\ (P_2 - P_1) \cdot \mathbf{d}_2 \end{pmatrix}$$

Resolviendo este sistema $2\times 2$, se obtienen los puntos $Q_1 = L_1(t^*)$ y $Q_2 = L_2(s^*)$, cumpliéndose:
$$\|Q_2 - Q_1\| = d(L_1, L_2)$$

## 2. Implementación Computacional

Modelado de la clase `RectaR3`, clasificación algorítmica de posiciones relativas, resolución del sistema de Gram $2	imes 2$ y renderizado 3D individual de alta fidelidad.


```
class RectaR3:
    """Representa una recta en R3 mediante r(t) = P0 + t * d."""
    def __init__(self, punto: np.ndarray, director: np.ndarray, nombre: str = "L"):
        self.p0 = np.array(punto, dtype=float)
        self.d = np.array(director, dtype=float)
        self.nombre = nombre
        if np.isclose(np.linalg.norm(self.d), 0.0):
            raise ValueError(f"El vector director de {nombre} no puede ser nulo.")
    
    def punto_en_parametro(self, t: float) -> np.ndarray:
        return self.p0 + t * self.d

    def ecuacion_vectorial(self) -> str:
        return f"{self.nombre}: (x, y, z) = ({self.p0[0]:g}, {self.p0[1]:g}, {self.p0[2]:g}) + t*({self.d[0]:g}, {self.d[1]:g}, {self.d[2]:g})"

    def ecuaciones_parametricas(self) -> str:
        p, d = self.p0, self.d
        sx = f"+ {d[0]:g}t" if d[0] >= 0 else f"- {abs(d[0]):g}t"
        sy = f"+ {d[1]:g}t" if d[1] >= 0 else f"- {abs(d[1]):g}t"
        sz = f"+ {d[2]:g}t" if d[2] >= 0 else f"- {abs(d[2]):g}t"
        return f"{self.nombre} => x = {p[0]:g} {sx},  y = {p[1]:g} {sy},  z = {p[2]:g} {sz}"


def calcular_puntos_mas_cercanos_alabeadas(r1: RectaR3, r2: RectaR3) -> tuple[np.ndarray, np.ndarray, float, float]:
    """Resuelve el sistema de Gram 2x2 para hallar Q1 y Q2."""
    p1, d1 = r1.p0, r1.d
    p2, d2 = r2.p0, r2.d
    delta_p = p2 - p1
    
    M = np.array([
        [np.dot(d1, d1), -np.dot(d1, d2)],
        [np.dot(d1, d2), -np.dot(d2, d2)]
    ], dtype=float)
    b = np.array([np.dot(delta_p, d1), np.dot(delta_p, d2)], dtype=float)
    
    sol = np.linalg.solve(M, b)
    t_opt, s_opt = float(sol[0]), float(sol[1])
    return r1.punto_en_parametro(t_opt), r2.punto_en_parametro(s_opt), t_opt, s_opt


def clasificar_posicion_relativa(r1: RectaR3, r2: RectaR3) -> dict:
    """Clasifica la relacion espacial de r1 y r2."""
    delta_p = r2.p0 - r1.p0
    cross_d = np.cross(r1.d, r2.d)
    norm_cross_d = float(np.linalg.norm(cross_d))
    
    if np.isclose(norm_cross_d, 0.0):
        cross_delta = np.cross(delta_p, r1.d)
        if np.isclose(np.linalg.norm(cross_delta), 0.0):
            return {'tipo': 'COINCIDENTES', 'distancia': 0.0}
        else:
            return {'tipo': 'PARALELAS', 'distancia': float(np.linalg.norm(cross_delta) / np.linalg.norm(r1.d))}
    
    pm = float(np.dot(delta_p, cross_d))
    if np.isclose(pm, 0.0, atol=1e-9):
        A = np.column_stack([r1.d, -r2.d])
        sol, _, _, _ = np.linalg.lstsq(A, delta_p, rcond=None)
        q = r1.punto_en_parametro(float(sol[0]))
        return {'tipo': 'SECANTES', 'distancia': 0.0, 'punto_interseccion': q}
    
    distancia = abs(pm) / norm_cross_d
    q1, q2, t_opt, s_opt = calcular_puntos_mas_cercanos_alabeadas(r1, r2)
    return {
        'tipo': 'ALABEADAS',
        'distancia': float(distancia),
        'q1': q1, 'q2': q2,
        't_opt': t_opt, 's_opt': s_opt
    }
```

```
# Definicion de rectas alabeadas (caso clasico)
r1 = RectaR3(punto=np.array([1.0, -2.0, 4.0]), director=np.array([2.0, 3.0, -1.0]), nombre="L1")
r2 = RectaR3(punto=np.array([0.0, 3.0, -1.0]), director=np.array([1.0, -1.0, 2.0]), nombre="L2")

print(r1.ecuacion_vectorial())
print(r1.ecuaciones_parametricas())
print(r2.ecuacion_vectorial())
print(r2.ecuaciones_parametricas())

res = clasificar_posicion_relativa(r1, r2)
print(f"Diagnostico: {res['tipo']}")
print(f"Distancia minima d(L1, L2): {res['distancia']:.4f} [u]")

if res['tipo'] == 'ALABEADAS':
    q1, q2 = res['q1'], res['q2']
    vec_q = q2 - q1
    print(f"Punto Q1 en L1 (t = {res['t_opt']:.4f}): {q1}")
    print(f"Punto Q2 en L2 (s = {res['s_opt']:.4f}): {q2}")
    print(f"Ortogonalidad con d1: {np.dot(vec_q, r1.d):.2e}")
    print(f"Ortogonalidad con d2: {np.dot(vec_q, r2.d):.2e}")
    print(f"Longitud ||Q2 - Q1||: {np.linalg.norm(vec_q):.4f}")
```

## 3. Discusión Geométrica de Resultados

1. **Dimensionalidad y Grados de Libertad:** En el plano bidimensional $\mathbb{R}^2$, dos rectas no coincidentes únicamente pueden ser paralelas o secantes. El espacio tridimensional $\mathbb{R}^3$ añade un grado de libertad que permite a las rectas cruzar sus trayectorias a diferentes profundidades sin intersecarse jamás ni ser paralelas; tal configuración define las **rectas alabeadas**.
2. **Propiedad de Separación entre Planos:** Toda pareja de rectas alabeadas puede encapsularse de manera unívoca en un par de planos paralelos $\Pi_1$ y $\Pi_2$ con vector normal unitario $\hat{\mathbf{n}} = \frac{\mathbf{d}_1 \times \mathbf{d}_2}{\|\mathbf{d}_1 \times \mathbf{d}_2\|}$. La distancia mínima entre las rectas equivale exactamente a la distancia perpendicular entre ambos planos envolventes.
3. **Unicidad del Segmento Común:** La condición de mínima distancia exige que el segmento $\vec{Q_1 Q_2}$ sea simultáneamente perpendicular a $\mathbf{d}_1$ y $\mathbf{d}_2$. La matriz del sistema lineal de Gram resultante tiene determinante no nulo $\|\mathbf{d}_1\|^2 \|\mathbf{d}_2\|^2 - (\mathbf{d}_1 \cdot \mathbf{d}_2)^2 = \|\mathbf{d}_1 \times \mathbf{d}_2\|^2 > 0$, lo que garantiza la existencia y unicidad de los puntos $Q_1$ y $Q_2$.


```
# Explorador Interactivo de Rectas en R3
def actualizar_recta_l2(x0=0.0, y0=3.0, z0=-1.0, dx=1.0, dy=-1.0, dz=2.0):
    r1_d = RectaR3(punto=np.array([1.0, -2.0, 4.0]), director=np.array([2.0, 3.0, -1.0]), nombre="L1")
    r2_d = RectaR3(punto=np.array([x0, y0, z0]), director=np.array([dx, dy, dz]), nombre="L2")
    
    diag = clasificar_posicion_relativa(r1_d, r2_d)
    
    fig = plt.figure(figsize=(7, 6), facecolor='white')
    ax = fig.add_subplot(1, 1, 1, projection='3d')
    ax.set_facecolor('white')
    
    t_vals = np.linspace(-2.5, 2.5, 30)
    p1 = np.array([r1_d.punto_en_parametro(t) for t in t_vals])
    p2 = np.array([r2_d.punto_en_parametro(t) for t in t_vals])
    
    ax.plot(p1[:, 0], p1[:, 1], p1[:, 2], color=USS_BLUE, linewidth=2.5, label="L1 fija")
    ax.plot(p2[:, 0], p2[:, 1], p2[:, 2], color=USS_ACCENT_BLUE, linewidth=2.5, label="L2 movil")
    
    if diag['tipo'] == 'ALABEADAS':
        q1_d, q2_d = diag['q1'], diag['q2']
        ax.plot([q1_d[0], q2_d[0]], [q1_d[1], q2_d[1]], [q1_d[2], q2_d[2]], color=USS_GOLD, linewidth=3.0)
        ax.scatter([q1_d[0], q2_d[0]], [q1_d[1], q2_d[1]], [q1_d[2], q2_d[2]], color=USS_GOLD, s=60)
    
    ax.set_title(f"Posicion: {diag['tipo']} | Distancia = {diag['distancia']:.3f} [u]",
                 fontsize=11, fontweight='bold', color=USS_BLUE)
    ax.view_init(elev=25, azim=-50)
    ax.legend(loc='upper left', fontsize=8)
    plt.tight_layout()
    plt.show()

interact(actualizar_recta_l2,
         x0=widgets.FloatSlider(value=0.0, min=-4.0, max=4.0, step=0.5, description='P2_x:'),
         y0=widgets.FloatSlider(value=3.0, min=-4.0, max=4.0, step=0.5, description='P2_y:'),
         z0=widgets.FloatSlider(value=-1.0, min=-4.0, max=4.0, step=0.5, description='P2_z:'),
         dx=widgets.FloatSlider(value=1.0, min=-3.0, max=3.0, step=0.5, description='d2_x:'),
         dy=widgets.FloatSlider(value=-1.0, min=-3.0, max=3.0, step=0.5, description='d2_y:'),
         dz=widgets.FloatSlider(value=2.0, min=-3.0, max=3.0, step=0.5, description='d2_z:'));
```

## 📚 1. Fundamentos Teóricos de las Rectas en $\mathbb{R}^3$

### 1.1. Ecuaciones de la Recta
Una recta $L$ en $\mathbb{R}^3$ queda unívocamente determinada por un punto base $P_0(x_0, y_0, z_0)$ y un vector director no nulo $\mathbf{d} = (d_x, d_y, d_z)$:

1. **Ecuación Vectorial:**
   $$\mathbf{r}(t) = P_0 + t\,\mathbf{d} = (x_0, y_0, z_0) + t(d_x, d_y, d_z), \quad t \in \mathbb{R}$$

2. **Ecuaciones Paramétricas:**
   $$\begin{cases} x(t) = x_0 + t\,d_x \\ y(t) = y_0 + t\,d_y \\ z(t) = z_0 + t\,d_z \end{cases}$$

3. **Ecuaciones Simétricas** (suponiendo $d_x, d_y, d_z \neq 0$):
   $$\frac{x - x_0}{d_x} = \frac{y - y_0}{d_y} = \frac{z - z_0}{d_z}$$

---

### 1.2. Clasificación de Posiciones Relativas de dos Rectas
Dadas $L_1: \mathbf{r}_1(t) = P_1 + t\,\mathbf{d}_1$ y $L_2: \mathbf{r}_2(s) = P_2 + s\,\mathbf{d}_2$:

```mermaid
flowchart TD
    A["¿d1 x d2 == 0?"] -->|SÍ: Paralelas| B["¿(P2 - P1) x d1 == 0?"]
    B -->|SÍ| C["COINCIDENTES (Misma Recta)"]
    B -->|NO| D["PARALELAS NO COINCIDENTES"]
    A -->|NO: No Paralelas| E["¿(P2 - P1) · (d1 x d2) == 0?"]
    E -->|SÍ: Coplanares| F["SECANTES (Punto Único)"]
    E -->|NO: No Coplanares| G["ALABEADAS / SKEW"]
```

---

### 1.3. Distancia Mínima entre Rectas Alabeadas
Dado que las rectas no se intersecan ni son paralelas, residen en dos planos paralelos $\Pi_1$ y $\Pi_2$ cuya normal común es $\mathbf{n} = \mathbf{d}_1 \times \mathbf{d}_2$. La distancia entre ambas es la proyección del vector $\vec{P_1 P_2}$ sobre dicha normal:
$$d(L_1, L_2) = \frac{|(P_2 - P_1) \cdot (\mathbf{d}_1 \times \mathbf{d}_2)|}{\|\mathbf{d}_1 \times \mathbf{d}_2\|}$$

> [!NOTE] Diferencia Dimensional entre $\mathbb{R}^2$ y $\mathbb{R}^3$
> En el plano $\mathbb{R}^2$, dos rectas que no se cortan son obligatoriamente paralelas. En el espacio tridimensional $\mathbb{R}^3$, surge un grado de libertad adicional: dos rectas pueden no intersecarse y no ser paralelas; en tal caso se denominan **alabeadas** (*skew lines*).

> [!IMPORTANT] Invarianza de la Distancia Mínima y Planos Paralelos
> La distancia mínima $d(L_1, L_2)$ entre dos rectas alabeadas es estrictamente positiva ($d > 0$) e invariante. Coincide exactamente con la distancia perpendicular entre los dos planos paralelos $\Pi_1$ y $\Pi_2$ que las contienen. El vector $\vec{Q_1 Q_2}$ es simultáneamente perpendicular a $\mathbf{d}_1$ y $\mathbf{d}_2$.

---

### 1.4. Determinación del Segmento Perpendicular Común
Buscamos los parámetros $t^*, s^*$ tales que el vector de enlace $\mathbf{w} = L_2(s) - L_1(t) = (P_2 - P_1) + s\mathbf{d}_2 - t\mathbf{d}_1$ sea ortogonal a ambos directores:
$$\begin{cases} \mathbf{w} \cdot \mathbf{d}_1 = 0 \\ \mathbf{w} \cdot \mathbf{d}_2 = 0 \end{cases} \implies \begin{pmatrix} \|\mathbf{d}_1\|^2 & -(\mathbf{d}_1 \cdot \mathbf{d}_2) \\ \mathbf{d}_1 \cdot \mathbf{d}_2 & -\|\mathbf{d}_2\|^2 \end{pmatrix} \begin{pmatrix} t \\ s \end{pmatrix} = \begin{pmatrix} (P_2 - P_1) \cdot \mathbf{d}_1 \\ (P_2 - P_1) \cdot \mathbf{d}_2 \end{pmatrix}$$
Resolviendo este sistema $2\times 2$, se obtienen los puntos de máxima cercanía $Q_1 = L_1(t^*)$ y $Q_2 = L_2(s^*)$, cumpliéndose estrictamente:
$$\|Q_2 - Q_1\| = d(L_1, L_2)$$


````python
# ==============================================================================
# 🛠️ CONFIGURACIÓN DEL ENTORNO Y PALETA INSTITUCIONAL USS
# ==============================================================================
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import ipywidgets as widgets
from ipywidgets import interact

USS_BLUE = '#00205B'          # Primario institucional (Recta L1)
USS_GOLD = '#D4AF37'          # Secundario dorado (Recta L2)
USS_ACCENT_BLUE = '#1E88E5'   # Azul de realce
USS_ACCENT_GREEN = '#27AE60'  # Verde de realce (puntos Q1, Q2)
USS_ACCENT_RED = '#C0392B'    # Rojo de realce (segmento mínimo)
USS_GRAY = '#7F8C8D'          # Gris auxiliar
USS_LIGHT_BG = '#F8F9FA'      # Fondo claro

print("✅ Entorno de rectas y planos configurado.")
````

````python
# ==============================================================================
# 📐 CLASE RECTA Y CÁLCULO DE PUNTOS MÁS CERCANOS
# ==============================================================================

class RectaR3:
    def __init__(self, p0, d, nombre="L"):
        self.p0 = np.array(p0, dtype=float)
        self.d = np.array(d, dtype=float)
        self.nombre = nombre
        if np.isclose(np.linalg.norm(self.d), 0.0):
            raise ValueError("El vector director no puede ser nulo.")

    def punto(self, t):
        return self.p0 + t * self.d

def analizar_rectas(r1, r2):
    p1, d1 = r1.p0, r1.d
    p2, d2 = r2.p0, r2.d
    delta_p = p2 - p1
    cross_d = np.cross(d1, d2)
    norm_cross = float(np.linalg.norm(cross_d))
    
    if np.isclose(norm_cross, 0.0):
        # Paralelas o Coincidentes
        cross_delta = np.cross(delta_p, d1)
        if np.isclose(np.linalg.norm(cross_delta), 0.0):
            return {'tipo': 'COINCIDENTES', 'distancia': 0.0, 'q1': p1, 'q2': p1}
        else:
            dist = float(np.linalg.norm(cross_delta) / np.linalg.norm(d1))
            return {'tipo': 'PARALELAS', 'distancia': dist, 'q1': p1, 'q2': p1}
            
    pm = float(np.dot(delta_p, cross_d))
    
    # Resolver sistema 2x2 para t y s del segmento ortogonal común
    M = np.array([
        [np.dot(d1, d1), -np.dot(d1, d2)],
        [np.dot(d1, d2), -np.dot(d2, d2)]
    ], dtype=float)
    b = np.array([np.dot(delta_p, d1), np.dot(delta_p, d2)], dtype=float)
    sol = np.linalg.solve(M, b)
    t_opt, s_opt = float(sol[0]), float(sol[1])
    
    q1 = r1.punto(t_opt)
    q2 = r2.punto(s_opt)
    distancia = float(np.linalg.norm(q2 - q1))
    
    tipo = 'SECANTES' if np.isclose(pm, 0.0, atol=1e-8) else 'ALABEADAS'
    return {
        'tipo': tipo,
        'distancia': distancia,
        'q1': q1, 'q2': q2,
        't_opt': t_opt, 's_opt': s_opt,
        'cross_d': cross_d, 'pm': pm
    }

print("✅ Algoritmo de clasificación y distancia cargado con éxito.")
````

````python
# ==============================================================================
# 🎮 SIMULADOR 3D INTERACTIVO: RECTAS Y SEGMENTO PERPENDICULAR COMÚN
# ==============================================================================

def simular_rectas_3d(p1x=1.0, p1y=-2.0, p1z=4.0,
                      d1x=2.0, d1y=3.0, d1z=-1.0,
                      p2x=0.0, p2y=3.0, p2z=-1.0,
                      d2x=1.0, d2y=-1.0, d2z=2.0):
    r1 = RectaR3([p1x, p1y, p1z], [d1x, d1y, d1z], "L1")
    r2 = RectaR3([p2x, p2y, p2z], [d2x, d2y, d2z], "L2")
    
    res = analizar_rectas(r1, r2)
    q1, q2 = res['q1'], res['q2']
    
    fig = plt.figure(figsize=(15, 6), facecolor='white')
    ax = fig.add_subplot(1, 2, 1, projection='3d')
    ax.set_title(f"Rectas en R³ — Posición: {res['tipo']}", fontsize=11, fontweight='bold', color=USS_BLUE)
    
    # Trazar rectas L1 y L2
    t_span = np.linspace(-3, 3, 50)
    l1_pts = np.array([r1.punto(t) for t in t_span])
    l2_pts = np.array([r2.punto(t) for t in t_span])
    
    ax.plot(l1_pts[:, 0], l1_pts[:, 1], l1_pts[:, 2], color=USS_BLUE, linewidth=3.0, label="L1")
    ax.plot(l2_pts[:, 0], l2_pts[:, 1], l2_pts[:, 2], color=USS_GOLD, linewidth=3.0, label="L2")
    
    # Puntos base y de acercamiento
    ax.scatter([r1.p0[0]], [r1.p0[1]], [r1.p0[2]], color=USS_BLUE, s=50, label="P1")
    ax.scatter([r2.p0[0]], [r2.p0[1]], [r2.p0[2]], color=USS_GOLD, s=50, label="P2")
    ax.scatter([q1[0]], [q1[1]], [q1[2]], color=USS_ACCENT_GREEN, s=70, marker='o', label="Q1")
    ax.scatter([q2[0]], [q2[1]], [q2[2]], color=USS_ACCENT_GREEN, s=70, marker='^', label="Q2")
    
    # Segmento perpendicular común
    ax.plot([q1[0], q2[0]], [q1[1], q2[1]], [q1[2], q2[2]], color=USS_ACCENT_RED, linewidth=4.0, label=f"Dist = {res['distancia']:.4f}")
    
    ax.set_xlabel('X', fontweight='bold', color=USS_BLUE)
    ax.set_ylabel('Y', fontweight='bold', color=USS_BLUE)
    ax.set_zlabel('Z', fontweight='bold', color=USS_BLUE)
    ax.legend(loc='upper left', fontsize=8)
    ax.view_init(elev=20, azim=60)
    
    # Panel analítico
    ax2 = fig.add_subplot(1, 2, 2)
    ax2.set_facecolor(USS_LIGHT_BG)
    ax2.set_title("Desglose Vectorial y Comprobaciones", fontsize=11, fontweight='bold', color=USS_BLUE)
    
    perp_v = q2 - q1
    dot1 = float(np.dot(perp_v, r1.d))
    dot2 = float(np.dot(perp_v, r2.d))
    
    info = (
        "📐 DATOS DE LAS RECTAS:\n"
        f" • L1: P1=({p1x:g}, {p1y:g}, {p1z:g}), d1=({d1x:g}, {d1y:g}, {d1z:g})\n"
        f" • L2: P2=({p2x:g}, {p2y:g}, {p2z:g}), d2=({d2x:g}, {d2y:g}, {d2z:g})\n\n"
        "🔍 DIAGNÓSTICO:\n"
        f" • Posición Relativa: {res['tipo']}\n"
        f" • Distancia Mínima = {res['distancia']:.4f} [u]\n\n"
        "📍 SEGMENTO PERPENDICULAR COMÚN:\n"
        f" • Parámetro t* en L1 = {res.get('t_opt', 0.0):.4f} => Q1 = ({q1[0]:.2f}, {q1[1]:.2f}, {q1[2]:.2f})\n"
        f" • Parámetro s* en L2 = {res.get('s_opt', 0.0):.4f} => Q2 = ({q2[0]:.2f}, {q2[1]:.2f}, {q2[2]:.2f})\n"
        f" • Vector (Q2 - Q1) = ({perp_v[0]:.2f}, {perp_v[1]:.2f}, {perp_v[2]:.2f})\n"
        f" • Verificación (Q2-Q1) · d1 = {dot1:.2e} (Ortogonal)\n"
        f" • Verificación (Q2-Q1) · d2 = {dot2:.2e} (Ortogonal)"
    )
    ax2.text(0.08, 0.92, info, transform=ax2.transAxes, fontsize=10.5, verticalalignment='top',
             fontfamily='monospace', bbox=dict(boxstyle='round,pad=0.8', facecolor='white', edgecolor=USS_BLUE, linewidth=2.0))
    ax2.axis('off')
    
    plt.tight_layout()
    plt.show()

interact(simular_rectas_3d,
         p1x=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='P1_x:'),
         p1y=widgets.FloatSlider(value=-2.0, min=-5.0, max=5.0, step=0.5, description='P1_y:'),
         p1z=widgets.FloatSlider(value=4.0, min=-5.0, max=5.0, step=0.5, description='P1_z:'),
         d1x=widgets.FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5, description='d1_x:'),
         d1y=widgets.FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description='d1_y:'),
         d1z=widgets.FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='d1_z:'),
         p2x=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='P2_x:'),
         p2y=widgets.FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description='P2_y:'),
         p2z=widgets.FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='P2_z:'),
         d2x=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='d2_x:'),
         d2y=widgets.FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='d2_y:'),
         d2z=widgets.FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5, description='d2_z:'));
````

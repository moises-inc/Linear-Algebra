## 📚 1. Fundamentos Teóricos de los Planos en $\mathbb{R}^3$

### 1.1. Ecuación Punto-Normal y Ecuación General
Un plano $\Pi$ en el espacio tridimensional queda determinado de manera única por un punto de apoyo $P_0(x_0, y_0, z_0)$ y un **vector normal** ortogonal a toda dirección contenida en el plano:
$$\mathbf{n} = (A, B, C) \neq \mathbf{0}$$

Para cualquier punto $P(x, y, z) \in \Pi$, el vector $\vec{P_0 P} = (x - x_0, y - y_0, z - z_0)$ debe ser ortogonal a $\mathbf{n}$:
$$\mathbf{n} \cdot (\mathbf{r} - P_0) = 0 \iff A(x - x_0) + B(y - y_0) + C(z - z_0) = 0$$

Desarrollando los términos y definiendo el término independiente $D = -(Ax_0 + By_0 + Cz_0)$, se obtiene la **ecuación cartesiana general**:
$$Ax + By + Cz + D = 0$$

---

### 1.2. Plano Determinado por Tres Puntos No Colineales
Dados $P_1, P_2, P_3 \in \mathbb{R}^3$, formamos los vectores concurrentes en $P_1$:
$$\mathbf{u} = P_2 - P_1, \quad \mathbf{v} = P_3 - P_1$$
El vector normal al plano es el producto cruz:
$$\mathbf{n} = \mathbf{u} \times \mathbf{v} = (P_2 - P_1) \times (P_3 - P_1)$$
Si $\|\mathbf{n}\| \neq 0$, el plano existe y es único.

---

### 1.3. Posiciones Relativas y Ángulo Diedro
Dadas las ecuaciones generales de dos planos:
$$\Pi_1: A_1 x + B_1 y + C_1 z + D_1 = 0, \quad \Pi_2: A_2 x + B_2 y + C_2 z + D_2 = 0$$
con normales $\mathbf{n}_1 = (A_1, B_1, C_1)$ y $\mathbf{n}_2 = (A_2, B_2, C_2)$:

1. **Paralelos o Coincidentes:** Si $\mathbf{n}_1 \times \mathbf{n}_2 = \mathbf{0}$ ($\mathbf{n}_1 = k\,\mathbf{n}_2$).
   - Si $D_1 = k\,D_2$ $\implies$ **Coincidentes**.
   - Si $D_1 \neq k\,D_2$ $\implies$ **Paralelos no coincidentes** (distancia $d = \frac{|D_1 - k D_2|}{\|\mathbf{n}_1\|}$).
2. **Secantes:** Si $\mathbf{n}_1 \times \mathbf{n}_2 \neq \mathbf{0}$. Se intersecan en una recta infinita común $L$.

El **ángulo diedro** $\theta \in [0, \pi/2]$ entre dos planos secantes es el ángulo entre sus vectores normales:
$$\cos\theta = \frac{|\mathbf{n}_1 \cdot \mathbf{n}_2|}{\|\mathbf{n}_1\| \|\mathbf{n}_2\|}$$

> [!IMPORTANT] Convención de Ángulo Diedro Agudo y Perpendicularidad
> Por convención trigonométrica y geométrica, el ángulo diedro se define siempre en el intervalo cerrado $[0^\circ, 90^\circ]$ ($[0, \pi/2]$ rad) mediante el valor absoluto del producto punto $|\mathbf{n}_1 \cdot \mathbf{n}_2|$. Dos planos son perpendiculares si y solo si $\mathbf{n}_1 \cdot \mathbf{n}_2 = 0$.

---

### 1.4. Recta de Intersección de dos Planos Secantes
La recta de intersección $L = \Pi_1 \cap \Pi_2$ cumple:
- **Vector director:** Como $L$ está contenida en $\Pi_1$ y en $\Pi_2$, debe ser perpendicular tanto a $\mathbf{n}_1$ como a $\mathbf{n}_2$:
  $$\mathbf{d} = \mathbf{n}_1 \times \mathbf{n}_2$$
- **Punto de paso $P_{\text{int}}$:** Se resuelve el sistema lineal subdeterminado de 2 ecuaciones con 3 incógnitas fijando una de las coordenadas en $0$ (seleccionando el eje con mayor componente en $\mathbf{d}$ para garantizar estabilidad numérica).

---

### 1.5. Distancia de un Punto $Q$ a un Plano $\Pi$
$$d(Q, \Pi) = \frac{|A x_q + B y_q + C z_q + D|}{\sqrt{A^2 + B^2 + C^2}}$$


```python
# ==============================================================================
# 🛠️ CONFIGURACIÓN DEL ENTORNO Y PALETA INSTITUCIONAL USS
# ==============================================================================
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import ipywidgets as widgets
from ipywidgets import interact

USS_BLUE = '#00205B'          # Primario institucional (Plano Pi1)
USS_GOLD = '#D4AF37'          # Secundario dorado (Plano Pi2 / Intersección)
USS_ACCENT_BLUE = '#1E88E5'   # Azul de realce (Normal n1)
USS_ACCENT_GREEN = '#27AE60'  # Verde de realce (Normal n2)
USS_ACCENT_RED = '#C0392B'    # Rojo de realce (Recta L)
USS_GRAY = '#7F8C8D'          # Gris auxiliar
USS_LIGHT_BG = '#F8F9FA'      # Fondo claro

print("✅ Entorno de planos e intersecciones configurado con Paleta USS.")
```

```python
# ==============================================================================
# 📐 CLASE PLANO R3 Y CÁLCULO DE LA RECTA DE INTERSECCIÓN
# ==============================================================================

class PlanoR3:
    def __init__(self, A, B, C, D, nombre="Pi"):
        self.A, self.B, self.C, self.D = float(A), float(B), float(C), float(D)
        self.nombre = nombre
        self.n = np.array([self.A, self.B, self.C], dtype=float)
        self.norma_n = float(np.linalg.norm(self.n))
        if np.isclose(self.norma_n, 0.0):
            raise ValueError("El vector normal no puede ser el vector nulo.")

    def evaluar(self, p):
        return float(np.dot(self.n, p) + self.D)

    def distancia_punto(self, q):
        return abs(self.evaluar(q)) / self.norma_n

def intersecar_planos(p1, p2):
    cross_n = np.cross(p1.n, p2.n)
    norm_cross = float(np.linalg.norm(cross_n))
    
    # Ángulo diedro
    cos_t = abs(np.dot(p1.n, p2.n)) / (p1.norma_n * p2.norma_n)
    cos_t = np.clip(cos_t, -1.0, 1.0)
    theta_deg = float(np.degrees(np.arccos(cos_t)))
    
    if np.isclose(norm_cross, 0.0):
        # Paralelos
        return {'tipo': 'PARALELOS', 'angulo_deg': 0.0, 'recta': None}
        
    # Secantes: hallar punto particular P_int fijando la variable con mayor d
    idx_fijo = int(np.argmax(np.abs(cross_n)))
    cols = [i for i in range(3) if i != idx_fijo]
    
    M_sub = np.array([
        [p1.n[cols[0]], p1.n[cols[1]]],
        [p2.n[cols[0]], p2.n[cols[1]]]
    ], dtype=float)
    b_sub = np.array([-p1.D, -p2.D], dtype=float)
    sol_sub = np.linalg.solve(M_sub, b_sub)
    
    p_int = np.zeros(3)
    p_int[cols[0]] = sol_sub[0]
    p_int[cols[1]] = sol_sub[1]
    p_int[idx_fijo] = 0.0
    
    return {
        'tipo': 'SECANTES',
        'angulo_deg': theta_deg,
        'vector_director': cross_n,
        'punto_paso': p_int,
        'err_p1': abs(p1.evaluar(p_int)),
        'err_p2': abs(p2.evaluar(p_int))
    }

print("✅ Algoritmo analítico de planos cargado con éxito.")
```

```python
# ==============================================================================
# 🎮 SIMULADOR 3D INTERACTIVO: PLANOS TRANSLÚCIDOS Y RECTA DE INTERSECCIÓN
# ==============================================================================

def simular_planos_3d(A1=2.0, B1=-1.0, C1=1.0, D1=-4.0,
                      A2=1.0, B2=2.0, C2=-1.0, D2=1.0):
    try:
        p1 = PlanoR3(A1, B1, C1, D1, "Pi1")
        p2 = PlanoR3(A2, B2, C2, D2, "Pi2")
    except ValueError as e:
        print(f"⚠️ Error: {e}")
        return
        
    res = intersecar_planos(p1, p2)
    
    fig = plt.figure(figsize=(15, 6), facecolor='white')
    ax = fig.add_subplot(1, 2, 1, projection='3d')
    ax.set_title(f"Intersección de Planos — Diedro: {res['angulo_deg']:.2f}°", fontsize=11, fontweight='bold', color=USS_BLUE)
    
    # Cuadrícula base
    centro = res['punto_paso'] if res['tipo'] == 'SECANTES' else np.array([0., 0., 0.])
    xr = np.linspace(centro[0] - 3, centro[0] + 3, 15)
    yr = np.linspace(centro[1] - 3, centro[1] + 3, 15)
    X, Y = np.meshgrid(xr, yr)
    
    # Graficar superficie de un plano
    def dibujar_plano(plano, color_surf):
        if not np.isclose(plano.C, 0.0):
            Z = (-plano.A * X - plano.B * Y - plano.D) / plano.C
            ax.plot_surface(X, Y, Z, color=color_surf, alpha=0.35, edgecolor='none')
        elif not np.isclose(plano.B, 0.0):
            zr = np.linspace(centro[2] - 3, centro[2] + 3, 15)
            X_m, Z_m = np.meshgrid(xr, zr)
            Y_m = (-plano.A * X_m - plano.D) / plano.B
            ax.plot_surface(X_m, Y_m, Z_m, color=color_surf, alpha=0.35, edgecolor='none')
            
    dibujar_plano(p1, USS_BLUE)
    dibujar_plano(p2, USS_GOLD)
    
    if res['tipo'] == 'SECANTES':
        p_int = res['punto_paso']
        d_dir = res['vector_director']
        t_span = np.linspace(-3.5, 3.5, 50)
        pts_l = np.array([p_int + t * d_dir for t in t_span])
        
        # Recta de intersección destacada
        ax.plot(pts_l[:, 0], pts_l[:, 1], pts_l[:, 2], color=USS_ACCENT_RED, linewidth=4.0, label="Recta L = Pi1 ∩ Pi2")
        ax.scatter([p_int[0]], [p_int[1]], [p_int[2]], color=USS_ACCENT_RED, s=60, label="P0 de paso")
        
        # Vectores normales
        n1_u = p1.n / p1.norma_n
        n2_u = p2.n / p2.norma_n
        ax.quiver(p_int[0], p_int[1], p_int[2], n1_u[0]*1.5, n1_u[1]*1.5, n1_u[2]*1.5, color=USS_ACCENT_BLUE, linewidth=2.5, label="n1")
        ax.quiver(p_int[0], p_int[1], p_int[2], n2_u[0]*1.5, n2_u[1]*1.5, n2_u[2]*1.5, color=USS_ACCENT_GREEN, linewidth=2.5, label="n2")
        
    ax.set_xlabel('X', fontweight='bold', color=USS_BLUE)
    ax.set_ylabel('Y', fontweight='bold', color=USS_BLUE)
    ax.set_zlabel('Z', fontweight='bold', color=USS_BLUE)
    ax.legend(loc='upper left', fontsize=8)
    ax.view_init(elev=25, azim=45)
    
    # Panel Analítico
    ax2 = fig.add_subplot(1, 2, 2)
    ax2.set_facecolor(USS_LIGHT_BG)
    ax2.set_title("Ecuaciones y Resultados Analíticos", fontsize=11, fontweight='bold', color=USS_BLUE)
    
    info = (
        "📐 PLANOS EN R³:\n"
        f" • Pi1: {A1:g}x + {B1:g}y + {C1:g}z + ({D1:g}) = 0\n"
        f"   Normal n1 = ({A1:g}, {B1:g}, {C1:g})\n\n"
        f" • Pi2: {A2:g}x + {B2:g}y + {C2:g}z + ({D2:g}) = 0\n"
        f"   Normal n2 = ({A2:g}, {B2:g}, {C2:g})\n\n"
        "🔍 RELACIÓN GEOMÉTRICA:\n"
        f" • Diagnóstico: {res['tipo']}\n"
        f" • Ángulo Diedro theta = {res['angulo_deg']:.2f}°\n\n"
    )
    if res['tipo'] == 'SECANTES':
        d = res['vector_director']
        p0 = res['punto_paso']
        info += (
            "📍 RECTA DE INTERSECCIÓN:\n"
            f" • Vector Director d = n1 x n2 = ({d[0]:.2f}, {d[1]:.2f}, {d[2]:.2f})\n"
            f" • Punto de Paso P0 = ({p0[0]:.2f}, {p0[1]:.2f}, {p0[2]:.2f})\n"
            f" • Verificación d · n1 = {np.dot(d, p1.n):.2e}\n"
            f" • Verificación d · n2 = {np.dot(d, p2.n):.2e}\n"
            f" • Pertenencia a Pi1: {res['err_p1']:.2e}\n"
            f" • Pertenencia a Pi2: {res['err_p2']:.2e}"
        )
        
    ax2.text(0.08, 0.92, info, transform=ax2.transAxes, fontsize=10.5, verticalalignment='top',
             fontfamily='monospace', bbox=dict(boxstyle='round,pad=0.8', facecolor='white', edgecolor=USS_GOLD, linewidth=2.0))
    ax2.axis('off')
    
    plt.tight_layout()
    plt.show()

interact(simular_planos_3d,
         A1=widgets.FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5, description='A1:'),
         B1=widgets.FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='B1:'),
         C1=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='C1:'),
         D1=widgets.FloatSlider(value=-4.0, min=-8.0, max=8.0, step=0.5, description='D1:'),
         A2=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='A2:'),
         B2=widgets.FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5, description='B2:'),
         C2=widgets.FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='C2:'),
         D2=widgets.FloatSlider(value=1.0, min=-8.0, max=8.0, step=0.5, description='D2:'));
```

## 📝 Resumen y Conclusiones del Módulo 07

> [!NOTE] Conexión Teórica con Sistemas de Ecuaciones y Teorema de Rouché-Frobenius
> La intersección de dos planos secantes equivale a resolver el sistema de $2 \times 3$:
> $$\begin{pmatrix} A_1 & B_1 & C_1 \\ A_2 & B_2 & C_2 \end{pmatrix} \begin{pmatrix} x \\ y \\ z \end{pmatrix} = \begin{pmatrix} -D_1 \\ -D_2 \end{pmatrix}$$
> Dado que $\mathbf{n}_1 \nparallel \mathbf{n}_2$, la matriz de coeficientes tiene $\operatorname{rango}(A) = 2$ y la ampliada $\operatorname{rango}(A|B) = 2$. Con $n = 3$ incógnitas:
> $$\text{Grados de libertad} = n - \operatorname{rango} = 3 - 2 = 1$$
> El conjunto solución tiene dimensión 1: geométricamente representa una **recta en el espacio $\mathbb{R}^3$**.

> [!EXAMPLE] Aplicación en Arquitectura, CAD y Computación Gráfica
> 1. **Diseño de Cubiertas y Techos (Aguas):** La arista de intersección entre dos faldones o pendientes de una cubierta (cumbrera o limatesa) se modela exactamente mediante la intersección de dos planos secantes en $\mathbb{R}^3$.
> 2. **Renderizado por Oclusión y Recorte (Frustum Clipping):** En motores de videojuegos, los 6 planos del volumen de visión (*camera frustum*) definen el espacio visible; la intersección entre estos planos delimita las aristas de la pirámide truncada.

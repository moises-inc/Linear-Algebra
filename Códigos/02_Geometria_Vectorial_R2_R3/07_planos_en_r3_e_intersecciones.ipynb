## 1. Fundamentos Teóricos

### 1.1. Ecuación Punto-Normal y Ecuación General
Un plano $\Pi$ en $\mathbb{R}^3$ queda determinado unívocamente por un punto de apoyo $P_0(x_0, y_0, z_0)$ y un **vector normal** ortogonal a cualquier dirección contenida en el plano:
$$\mathbf{n} = (A, B, C) \neq \mathbf{0}$$

Para cualquier punto genérico $P(x, y, z) \in \Pi$, el vector $\vec{P_0 P} = (x - x_0, y - y_0, z - z_0)$ es ortogonal a $\mathbf{n}$:
$$\mathbf{n} \cdot (\mathbf{r} - P_0) = 0 \iff A(x - x_0) + B(y - y_0) + C(z - z_0) = 0$$

Definiendo el término independiente $D = -(A x_0 + B y_0 + C z_0)$, se obtiene la **ecuación cartesiana general**:
$$Ax + By + Cz + D = 0$$

---

### 1.2. Plano Determinado por Tres Puntos No Colineales
Dados tres puntos $P_1, P_2, P_3 \in \mathbb{R}^3$ no alineados, los vectores concurrentes son $\mathbf{u} = P_2 - P_1$ y $\mathbf{v} = P_3 - P_1$. El vector normal se obtiene mediante el producto cruz:
$$\mathbf{n} = (P_2 - P_1) \times (P_3 - P_1)$$
Si $\|\mathbf{n}\| \neq 0$, el plano existe y es único.

---

### 1.3. Posiciones Relativas y Ángulo Diedro
Dadas las ecuaciones generales de dos planos:
$$\Pi_1: A_1 x + B_1 y + C_1 z + D_1 = 0, \quad \Pi_2: A_2 x + B_2 y + C_2 z + D_2 = 0$$
con vectores normales $\mathbf{n}_1 = (A_1, B_1, C_1)$ y $\mathbf{n}_2 = (A_2, B_2, C_2)$:

- **Paralelos o Coincidentes:** Si $\mathbf{n}_1 \times \mathbf{n}_2 = \mathbf{0}$.
  - Coincidentes si sus ecuaciones son múltiplos proporcionales ($D_2 = k D_1$).
  - Paralelos no coincidentes si los términos independientes difieren.
- **Secantes:** Si $\mathbf{n}_1 \times \mathbf{n}_2 \neq \mathbf{0}$. Se intersecan en una recta infinita común $L$.

El **ángulo diedro agudo** $\theta \in [0, \pi/2]$ entre dos planos secantes se define como:
$$\cos\theta = \frac{|\mathbf{n}_1 \cdot \mathbf{n}_2|}{\|\mathbf{n}_1\| \|\mathbf{n}_2\|}$$

> [!IMPORTANT] Perpendicularidad de Planos
> Dos planos son perpendiculares entre sí si y solo si sus vectores normales son ortogonales:
> $$\Pi_1 \perp \Pi_2 \iff \mathbf{n}_1 \cdot \mathbf{n}_2 = 0$$

---

### 1.4. Recta de Intersección de Planos Secantes
La recta de intersección $L = \Pi_1 \cap \Pi_2$ cumple:
- **Vector director:** Como $L$ reside en ambos planos, es simultáneamente ortogonal a $\mathbf{n}_1$ y $\mathbf{n}_2$:
  $$\mathbf{d} = \mathbf{n}_1 \times \mathbf{n}_2$$
- **Punto de paso $P_0$:** Se obtiene fijando en cero la coordenada correspondiente a la componente de mayor valor absoluto en $\mathbf{d}$ (pivoteo numérico) y resolviendo el sistema $2\times 2$ restante.

---

### 1.5. Distancia de un Punto a un Plano y Proyección Ortogonal
Dada la posición $Q(x_q, y_q, z_q)$ y el plano $\Pi: Ax + By + Cz + D = 0$:
$$d(Q, \Pi) = \frac{|A x_q + B y_q + C z_q + D|}{\sqrt{A^2 + B^2 + C^2}}$$

El punto proyección ortogonal $Q_{\text{proj}} \in \Pi$ más cercano a $Q$ es:
$$Q_{\text{proj}} = Q - \left(\frac{A x_q + B y_q + C z_q + D}{\|\mathbf{n}\|^2}\right) \mathbf{n}$$

## 2. Implementación Computacional

Modelado de la clase `PlanoR3`, clasificación algorítmica de posiciones relativas, cálculo de intersección con pivoteo y renderizado de dos figuras 3D individuales de alta fidelidad.


```
class PlanoR3:
    """Representa un plano en R3 mediante Ax + By + Cz + D = 0."""
    def __init__(self, A: float, B: float, C: float, D: float, nombre: str = "Pi"):
        self.A, self.B, self.C, self.D = float(A), float(B), float(C), float(D)
        self.nombre = nombre
        self.n = np.array([self.A, self.B, self.C], dtype=float)
        self.norma_n = float(np.linalg.norm(self.n))
        if np.isclose(self.norma_n, 0.0):
            raise ValueError(f"El vector normal de {nombre} no puede ser nulo.")

    @classmethod
    def desde_punto_y_normal(cls, p0: np.ndarray, normal: np.ndarray, nombre: str = "Pi"):
        p0, n = np.array(p0, dtype=float), np.array(normal, dtype=float)
        return cls(n[0], n[1], n[2], -float(np.dot(n, p0)), nombre=nombre)

    @classmethod
    def desde_tres_puntos(cls, p1: np.ndarray, p2: np.ndarray, p3: np.ndarray, nombre: str = "Pi"):
        v1 = np.array(p2, dtype=float) - np.array(p1, dtype=float)
        v2 = np.array(p3, dtype=float) - np.array(p1, dtype=float)
        normal = np.cross(v1, v2)
        if np.isclose(np.linalg.norm(normal), 0.0):
            raise ValueError("Puntos colineales.")
        return cls.desde_punto_y_normal(p1, normal, nombre=nombre)

    def punto_arbitrario(self) -> np.ndarray:
        idx = int(np.argmax(np.abs(self.n)))
        p = np.zeros(3)
        p[idx] = -self.D / self.n[idx]
        return p

    def ecuacion_general(self) -> str:
        return f"{self.nombre}: {self.A:g}x + {self.B:g}y + {self.C:g}z + {self.D:g} = 0"

    def evaluar_punto(self, punto: np.ndarray) -> float:
        return float(np.dot(self.n, punto) + self.D)

    def distancia_punto(self, punto: np.ndarray) -> float:
        return abs(self.evaluar_punto(punto)) / self.norma_n

    def proyeccion_punto(self, punto: np.ndarray) -> np.ndarray:
        t = -self.evaluar_punto(punto) / (self.norma_n**2)
        return punto + t * self.n


def angulo_diedro(p1: PlanoR3, p2: PlanoR3) -> tuple[float, float]:
    cos_theta = abs(float(np.dot(p1.n, p2.n))) / (p1.norma_n * p2.norma_n)
    theta_rad = float(np.arccos(np.clip(cos_theta, -1.0, 1.0)))
    return theta_rad, float(np.degrees(theta_rad))


def clasificar_e_intersecar_planos(p1: PlanoR3, p2: PlanoR3) -> dict:
    cross_n = np.cross(p1.n, p2.n)
    norm_cross = float(np.linalg.norm(cross_n))
    theta_rad, theta_deg = angulo_diedro(p1, p2)
    
    if np.isclose(norm_cross, 0.0):
        k = p2.norma_n / p1.norma_n
        if np.dot(p1.n, p2.n) < 0:
            k = -k
        if np.isclose(p2.D - k * p1.D, 0.0):
            return {'tipo': 'COINCIDENTES', 'angulo_deg': 0.0}
        else:
            return {'tipo': 'PARALELOS', 'distancia': float(p1.distancia_punto(p2.punto_arbitrario())), 'angulo_deg': 0.0}
    
    # Secantes: pivoteo numerico
    d = cross_n
    idx_fijo = int(np.argmax(np.abs(d)))
    cols = [i for i in range(3) if i != idx_fijo]
    M_sub = np.array([[p1.n[cols[0]], p1.n[cols[1]]], [p2.n[cols[0]], p2.n[cols[1]]]], dtype=float)
    b_sub = np.array([-p1.D, -p2.D], dtype=float)
    sol = np.linalg.solve(M_sub, b_sub)
    
    p0 = np.zeros(3)
    p0[cols[0]], p0[cols[1]] = sol[0], sol[1]
    
    return {
        'tipo': 'SECANTES', 'angulo_deg': theta_deg, 'angulo_rad': theta_rad,
        'vector_director': d, 'punto_paso': p0,
        'err_p1': abs(p1.evaluar_punto(p0)), 'err_p2': abs(p2.evaluar_punto(p0))
    }
```

```
# Casos numericos y verificaciones
p1 = PlanoR3(A=2.0, B=-1.0, C=1.0, D=-4.0, nombre="Pi1")
p2 = PlanoR3(A=1.0, B=2.0, C=-1.0, D=1.0, nombre="Pi2")

print(p1.ecuacion_general())
print(p2.ecuacion_general())

res = clasificar_e_intersecar_planos(p1, p2)
print(f"Diagnostico: {res['tipo']}")
print(f"Angulo Diedro: {res['angulo_deg']:.2f} deg ({res['angulo_rad']:.4f} rad)")

if res['tipo'] == 'SECANTES':
    d = res['vector_director']
    p0 = res['punto_paso']
    print(f"Vector director de la recta d = {d}")
    print(f"Punto particular P0 = {p0}")
    print(f"Verificacion pertenencia Pi1: {res['err_p1']:.2e}")
    print(f"Verificacion pertenencia Pi2: {res['err_p2']:.2e}")
    print(f"Ortogonalidad d . n1: {np.dot(d, p1.n):.2e}")
    print(f"Ortogonalidad d . n2: {np.dot(d, p2.n):.2e}")
```

## 3. Discusión Geométrica de Resultados

1. **Subespacios Afines e Intersección:** Un plano en $\mathbb{R}^3$ es una variedad lineal afín de dimensión 2 (codimensión 1). La intersección de dos planos no paralelos equivale a la solución de un sistema algebraico linealmente independiente de rango 2 con 3 incógnitas, resultando en un conjunto solución uniparamétrico de dimensión $3 - 2 = 1$, el cual describe geométricamente una recta.
2. **Estabilidad Numérica del Pivoteo:** Al despejar un punto particular $P_0$ sobre la recta de intersección, fijar arbitrariamente $z=0$ puede generar singularidades si la componente $d_z = (\mathbf{n}_1 \times \mathbf{n}_2)_z$ es nula (el plano no corta al eje Z). El pivoteo por máxima magnitud absoluta selecciona $\max(|d_x|, |d_y|, |d_z|)$, garantizando un determinante de submatriz máximo y evitando cualquier división por cero o error de precisión flotante.
3. **Métrica de Mínima Distancia y Ortogonalidad:** El vector diferencia $\vec{Q_{\text{proj}} Q} = Q - Q_{\text{proj}}$ es estrictamente colineal con el vector normal $\mathbf{n}$. Por consiguiente, cualquier otro punto $P \in \Pi$ forma un triángulo rectángulo con hipotenusa $\|Q - P\| > \|Q - Q_{\text{proj}}\|$, demostrando que la proyección ortogonal minimiza unívocamente la distancia euclidiana del punto al plano.


```
# Explorador Interactivo de Posiciones de Planos
def actualizar_planos(A2=1.0, B2=2.0, C2=-1.0, D2=1.0):
    p1 = PlanoR3(A=2.0, B=-1.0, C=1.0, D=-4.0, nombre="Pi1")
    if np.isclose(A2**2 + B2**2 + C2**2, 0.0):
        print("El vector normal no puede ser nulo.")
        return
    p2 = PlanoR3(A=A2, B=B2, C=C2, D=D2, nombre="Pi2")
    res = clasificar_e_intersecar_planos(p1, p2)
    
    fig = plt.figure(figsize=(7, 6), facecolor='white')
    ax = fig.add_subplot(1, 1, 1, projection='3d')
    ax.set_facecolor('white')
    
    xr = np.linspace(-3, 3, 12)
    yr = np.linspace(-3, 3, 12)
    X, Y = np.meshgrid(xr, yr)
    
    if not np.isclose(p1.C, 0.0):
        ax.plot_surface(X, Y, (-p1.A*X - p1.B*Y - p1.D)/p1.C, color=USS_BLUE, alpha=0.25)
    if not np.isclose(p2.C, 0.0):
        ax.plot_surface(X, Y, (-p2.A*X - p2.B*Y - p2.D)/p2.C, color=USS_GOLD, alpha=0.25)
    
    ax.set_title(f"Posicion: {res['tipo']} | Angulo = {res['angulo_deg']:.1f} deg",
                 fontsize=11, fontweight='bold', color=USS_BLUE)
    ax.view_init(elev=25, azim=-50)
    plt.tight_layout()
    plt.show()

interact(actualizar_planos,
         A2=widgets.FloatSlider(value=1.0, min=-3.0, max=3.0, step=0.5, description='A2:'),
         B2=widgets.FloatSlider(value=2.0, min=-3.0, max=3.0, step=0.5, description='B2:'),
         C2=widgets.FloatSlider(value=-1.0, min=-3.0, max=3.0, step=0.5, description='C2:'),
         D2=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='D2:'));
```

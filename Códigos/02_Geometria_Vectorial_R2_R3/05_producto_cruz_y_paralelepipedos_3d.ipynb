## 1. Fundamentos Teóricos

### 1.1. Producto Cruz (Vectorial)
A diferencia del producto punto (cuyo recorrido es el cuerpo de escalares $\mathbb{R}$), el **producto cruz** opera exclusivamente en $\mathbb{R}^3$ y genera un nuevo vector perpendicular a los dos vectores de entrada:
$$\mathbf{u} \times \mathbf{v} = \begin{vmatrix} \hat{\mathbf{i}} & \hat{\mathbf{j}} & \hat{\mathbf{k}} \\ u_1 & u_2 & u_3 \\ v_1 & v_2 & v_3 \end{vmatrix} = (u_2 v_3 - u_3 v_2)\hat{\mathbf{i}} - (u_1 v_3 - u_3 v_1)\hat{\mathbf{j}} + (u_1 v_2 - u_2 v_1)\hat{\mathbf{k}}$$

---

### 1.2. Propiedades Fundamentales e Identidad de Lagrange
1. **Anticonmutatividad:**
   $$\mathbf{v} \times \mathbf{u} = -(\mathbf{u} \times \mathbf{v})$$
2. **Ortogonalidad Simultánea Estricta:**
   $$(\mathbf{u} \times \mathbf{v}) \cdot \mathbf{u} = 0 \quad \text{y} \quad (\mathbf{u} \times \mathbf{v}) \cdot \mathbf{v} = 0$$
3. **Norma y Seno del Ángulo Subtendido:**
   $$\|\mathbf{u} \times \mathbf{v}\| = \|\mathbf{u}\| \|\mathbf{v}\| \sin\theta$$
4. **Identidad de Lagrange:**
   $$\|\mathbf{u} \times \mathbf{v}\|^2 = \|\mathbf{u}\|^2 \|\mathbf{v}\|^2 - (\mathbf{u} \cdot \mathbf{v})^2$$

> [!IMPORTANT] Interpretación Geométrica de Áreas en $\mathbb{R}^3$
> - **Área del Paralelogramo:** Sustentado por las aristas concurrentes $\mathbf{u}$ y $\mathbf{v}$:
>   $$A_{\text{paral}} = \|\mathbf{u} \times \mathbf{v}\|$$
> - **Área del Triángulo Sustentado:** Formado por $\mathbf{u}, \mathbf{v}$ y la cuerda que une sus extremos:
>   $$A_{\text{tri}} = \frac{1}{2} \|\mathbf{u} \times \mathbf{v}\|$$

---

### 1.3. Triple Producto Escalar (Producto Mixto) y Volúmenes
Dados tres vectores concurrentes $\mathbf{u}, \mathbf{v}, \mathbf{w} \in \mathbb{R}^3$, el producto mixto se define algebraicamente como el determinante de la matriz formada por sus componentes:
$$[\mathbf{u}, \mathbf{v}, \mathbf{w}] = \mathbf{u} \cdot (\mathbf{v} \times \mathbf{w}) = \det \begin{pmatrix} u_1 & u_2 & u_3 \\ v_1 & v_2 & v_3 \\ w_1 & w_2 & w_3 \end{pmatrix}$$

> [!NOTE] Interpretación de Volúmenes y Coplanaridad
> - **Volumen del Paralelepípedo:**
>   $$V_{\text{paral}} = |[\mathbf{u}, \mathbf{v}, \mathbf{w}]|$$
> - **Volumen del Tetraedro:**
>   $$V_{\text{tet}} = \frac{1}{6} |[\mathbf{u}, \mathbf{v}, \mathbf{w}]|$$
> - **Criterio de Coplanaridad:** $\mathbf{u}, \mathbf{v}, \mathbf{w}$ son coplanares (linealmente dependientes) si y solo si $[\mathbf{u}, \mathbf{v}, \mathbf{w}] = 0$.

## 2. Implementación Computacional

Implementación analítica del producto vectorial y mixto, demostración simbólica con SymPy y generación de dos figuras 3D independientes de alta fidelidad.


```
def producto_cruz_numerico(u: np.ndarray, v: np.ndarray) -> np.ndarray:
    """Calcula el producto cruz en R3."""
    return np.cross(u, v)

def producto_cruz_simbolico(u_sym: sp.Matrix, v_sym: sp.Matrix) -> sp.Matrix:
    """Calcula el producto cruz formal por cofactores."""
    cx = u_sym[1] * v_sym[2] - u_sym[2] * v_sym[1]
    cy = -(u_sym[0] * v_sym[2] - u_sym[2] * v_sym[0])
    cz = u_sym[0] * v_sym[1] - u_sym[1] * v_sym[0]
    return sp.Matrix([cx, cy, cz])

def verificar_ortogonalidad_producto_cruz(u: np.ndarray, v: np.ndarray) -> dict:
    """Comprueba que u x v sea perpendicular a u y a v."""
    c = producto_cruz_numerico(u, v)
    dot_u = float(np.dot(c, u))
    dot_v = float(np.dot(c, v))
    return {
        'cross': c,
        'dot_with_u': dot_u,
        'dot_with_v': dot_v,
        'ortogonal_a_u': bool(abs(dot_u) < 1e-12),
        'ortogonal_a_v': bool(abs(dot_v) < 1e-12)
    }

def verificar_identidad_lagrange(u: np.ndarray, v: np.ndarray) -> dict:
    """Verifica ||u x v||^2 = ||u||^2 ||v||^2 - (u . v)^2."""
    c = producto_cruz_numerico(u, v)
    lhs = float(np.linalg.norm(c)**2)
    rhs = float((np.linalg.norm(u)**2) * (np.linalg.norm(v)**2) - (np.dot(u, v)**2))
    diff = abs(lhs - rhs)
    return {'norm_cross_sq': lhs, 'lagrange_rhs': rhs, 'diferencia': diff, 'valido': bool(diff < 1e-11)}

def area_paralelogramo_y_triangulo(u: np.ndarray, v: np.ndarray) -> tuple[float, float]:
    """Calcula areas de paralelogramo y triangulo."""
    norm_c = float(np.linalg.norm(producto_cruz_numerico(u, v)))
    return norm_c, 0.5 * norm_c

def producto_mixto(u: np.ndarray, v: np.ndarray, w: np.ndarray) -> float:
    """Calcula [u, v, w] = det([u, v, w])."""
    return float(np.linalg.det(np.array([u, v, w], dtype=float)))

def volumen_paralelepipedo_y_tetraedro(u: np.ndarray, v: np.ndarray, w: np.ndarray) -> tuple[float, float]:
    """Calcula volumen de paralelepipedo y tetraedro."""
    vol_p = abs(producto_mixto(u, v, w))
    return float(vol_p), float(vol_p / 6.0)

def generar_vertices_paralelepipedo(u: np.ndarray, v: np.ndarray, w: np.ndarray,
                                     origen: np.ndarray = np.array([0.0, 0.0, 0.0])) -> dict:
    """Construye los 8 vertices y 6 caras del paralelepipedo."""
    o = np.array(origen, dtype=float)
    v0 = o
    v1, v2, v3 = o + u, o + u + v, o + v
    v4, v5, v6, v7 = o + w, o + u + w, o + u + v + w, o + v + w
    vertices = np.array([v0, v1, v2, v3, v4, v5, v6, v7])
    caras = [
        [v0, v1, v2, v3], [v4, v5, v6, v7],
        [v0, v1, v5, v4], [v2, v3, v7, v6],
        [v0, v3, v7, v4], [v1, v2, v6, v5]
    ]
    return {'vertices': vertices, 'caras': caras}
```

```
# Demostracion analitica formal con SymPy
u1, u2, u3 = sp.symbols('u1 u2 u3', real=True)
v1, v2, v3 = sp.symbols('v1 v2 v3', real=True)

u_s = sp.Matrix([u1, u2, u3])
v_s = sp.Matrix([v1, v2, v3])

cross_s = producto_cruz_simbolico(u_s, v_s)
dot_u_s = sp.simplify(cross_s.dot(u_s))
dot_v_s = sp.simplify(cross_s.dot(v_s))

print("Producto cruz u x v simbolico:")
display(cross_s)
print(f"Ortogonalidad (u x v) . u = {dot_u_s}")
print(f"Ortogonalidad (u x v) . v = {dot_v_s}")
assert dot_u_s == 0 and dot_v_s == 0

# Anticonmutatividad
cross_vu_s = producto_cruz_simbolico(v_s, u_s)
assert sp.simplify(cross_s + cross_vu_s) == sp.Matrix([0, 0, 0])
print("Anticonmutatividad verificada: u x v = -(v x u)")
```

```
# Evaluacion numerica
u = np.array([3.0, 0.0, 1.0])
v = np.array([1.0, 2.0, 0.0])
w = np.array([0.0, 1.0, 4.0])

c = producto_cruz_numerico(u, v)
ortog = verificar_ortogonalidad_producto_cruz(u, v)
lagrange = verificar_identidad_lagrange(u, v)
area_p, area_t = area_paralelogramo_y_triangulo(u, v)
vol_p, vol_t = volumen_paralelepipedo_y_tetraedro(u, v, w)

print(f"u = {u}, v = {v}, w = {w}")
print(f"u x v = {c}, ||u x v|| = {np.linalg.norm(c):.4f}")
print(f"Ortogonalidad u . (u x v) = {ortog['dot_with_u']:.2e}, v . (u x v) = {ortog['dot_with_v']:.2e}")
print(f"Identidad de Lagrange satisfecha: {lagrange['valido']} (Error: {lagrange['diferencia']:.2e})")
print(f"Area Paralelogramo: {area_p:.4f} [u^2] | Area Triangulo: {area_t:.4f} [u^2]")
print(f"Volumen Paralelepipedo: {vol_p:.4f} [u^3] | Volumen Tetraedro: {vol_t:.4f} [u^3]")
```

## 3. Discusión Geométrica de Resultados

1. **Ortogonalidad y Espacio Dual:** El producto vectorial $\mathbf{u} \times \mathbf{v}$ define el vector normal canónico al subespacio bidimensional generado por $\{\mathbf{u}, \mathbf{v}\}$. La doble condición de ortogonalidad $(\mathbf{u} \times \mathbf{v}) \cdot \mathbf{u} = 0$ y $(\mathbf{u} \times \mathbf{v}) \cdot \mathbf{v} = 0$ garantiza que dicho vector es perpendicular a cualquier combinación lineal en dicho plano.
2. **Interpretación Métrica del Determinante:** El triple producto escalar $[\mathbf{u}, \mathbf{v}, \mathbf{w}]$ proporciona el volumen con signo del paralelepípedo. El signo del determinante refleja la quiralidad u orientación de la terna de vectores (positiva para ternas diestras conforme a la regla de la mano derecha, negativa para ternas zurdas).
3. **Identidad de Lagrange:** Vincula el producto escalar con la norma del producto vectorial. Mientras que el producto punto captura la proyección colineal ($\cos\theta$), el producto cruz cuantifica la divergencia ortogonal ($\sin\theta$), cumpliendo $\|\mathbf{u} \times \mathbf{v}\|^2 + (\mathbf{u} \cdot \mathbf{v})^2 = \|\mathbf{u}\|^2 \|\mathbf{v}\|^2$.


```
# Explorador Interactivo del Producto Cruz
def actualizar_cruz(ux=3.0, uy=0.0, uz=1.0, vx=1.0, vy=2.0, vz=0.0):
    u_dyn = np.array([ux, uy, uz], dtype=float)
    v_dyn = np.array([vx, vy, vz], dtype=float)
    c_dyn = np.cross(u_dyn, v_dyn)
    norm_c_dyn = float(np.linalg.norm(c_dyn))
    
    fig = plt.figure(figsize=(7, 6), facecolor='white')
    ax = fig.add_subplot(1, 1, 1, projection='3d')
    ax.set_facecolor('white')
    
    cara = [np.zeros(3), u_dyn, u_dyn + v_dyn, v_dyn]
    poly = Poly3DCollection([cara], alpha=0.25, facecolor=USS_GOLD, edgecolor=USS_BLUE, linewidths=1.2)
    ax.add_collection3d(poly)
    
    ax.quiver(0, 0, 0, u_dyn[0], u_dyn[1], u_dyn[2], color=USS_BLUE, linewidth=2.5, label="u")
    ax.quiver(0, 0, 0, v_dyn[0], v_dyn[1], v_dyn[2], color=USS_GOLD, linewidth=2.5, label="v")
    ax.quiver(0, 0, 0, c_dyn[0], c_dyn[1], c_dyn[2], color=USS_ACCENT_RED, linewidth=2.5, label="u x v")
    
    ax.set_title(f"Area Paralelogramo = {norm_c_dyn:.2f} [u^2]", fontsize=11, fontweight='bold', color=USS_BLUE)
    ax.view_init(elev=25, azim=-50)
    ax.legend(loc='upper left', fontsize=8)
    plt.tight_layout()
    plt.show()

interact(actualizar_cruz,
         ux=widgets.FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description='u_x:'),
         uy=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='u_y:'),
         uz=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='u_z:'),
         vx=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='v_x:'),
         vy=widgets.FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5, description='v_y:'),
         vz=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='v_z:'));
```

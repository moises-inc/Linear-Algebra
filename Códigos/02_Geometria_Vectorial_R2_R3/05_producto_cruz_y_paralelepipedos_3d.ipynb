## 📚 1. Fundamentos Teóricos del Producto Vectorial y Mixto

### 1.1. Producto Cruz (Vectorial)
A diferencia del producto punto (cuyo resultado es un escalar), el **producto cruz** opera exclusivamente en $\mathbb{R}^3$ y produce un nuevo **vector** perpendicular a ambos:
$$\mathbf{u} \times \mathbf{v} = \begin{vmatrix} \hat{\mathbf{i}} & \hat{\mathbf{j}} & \hat{\mathbf{k}} \\ u_1 & u_2 & u_3 \\ v_1 & v_2 & v_3 \end{vmatrix} = (u_2 v_3 - u_3 v_2)\,\hat{\mathbf{i}} - (u_1 v_3 - u_3 v_1)\,\hat{\mathbf{j}} + (u_1 v_2 - u_2 v_1)\,\hat{\mathbf{k}}$$

---

### 1.2. Propiedades Fundamentales
1. **Anticonmutatividad:** $\mathbf{v} \times \mathbf{u} = -(\mathbf{u} \times \mathbf{v})$.
2. **Ortogonalidad Estricta:** $(\mathbf{u} \times \mathbf{v}) \cdot \mathbf{u} = 0 \quad \text{y} \quad (\mathbf{u} \times \mathbf{v}) \cdot \mathbf{v} = 0$.
3. **Norma y Seno del Ángulo:** $\|\mathbf{u} \times \mathbf{v}\| = \|\mathbf{u}\| \|\mathbf{v}\| \sin\theta$.
4. **Identidad de Lagrange:**
$$\|\mathbf{u} \times \mathbf{v}\|^2 = \|\mathbf{u}\|^2 \|\mathbf{v}\|^2 - (\mathbf{u} \cdot \mathbf{v})^2$$

> [!IMPORTANT] Interpretación Geométrica de Áreas
> - **Área del Paralelogramo:** Determinado por las aristas $\mathbf{u}$ y $\mathbf{v}$:
>   $$A_{\text{paral}} = \|\mathbf{u} \times \mathbf{v}\|$$
> - **Área del Triángulo Sustentado:** Formado por $\mathbf{u}, \mathbf{v}$ y el segmento que une sus extremos:
>   $$A_{\text{tri}} = \frac{1}{2}\|\mathbf{u} \times \mathbf{v}\|$$

---

### 1.3. Triple Producto Escalar (Producto Mixto) y Volúmenes
Dados tres vectores $\mathbf{u}, \mathbf{v}, \mathbf{w} \in \mathbb{R}^3$, su producto mixto se define como:
$$[\mathbf{u}, \mathbf{v}, \mathbf{w}] = \mathbf{u} \cdot (\mathbf{v} \times \mathbf{w}) = \det \begin{pmatrix} u_1 & u_2 & u_3 \\ v_1 & v_2 & v_3 \\ w_1 & w_2 & w_3 \end{pmatrix}$$

> [!NOTE] Interpretación de Volúmenes
> - **Volumen del Paralelepípedo:** Sustentado por las 3 aristas concurrentes $\mathbf{u}, \mathbf{v}, \mathbf{w}$:
>   $$V_{\text{paral}} = |[\mathbf{u}, \mathbf{v}, \mathbf{w}]|$$
> - **Volumen del Tetraedro:** Formado por los mismos 3 vectores:
>   $$V_{\text{tet}} = \frac{1}{6} |[\mathbf{u}, \mathbf{v}, \mathbf{w}]|$$
> - **Criterio de Coplanaridad:** $\mathbf{u}, \mathbf{v}, \mathbf{w}$ yacen en un mismo plano si y solo si $[\mathbf{u}, \mathbf{v}, \mathbf{w}] = 0$.


```python
# ==============================================================================
# 🛠️ CONFIGURACIÓN DEL ENTORNO Y PALETA INSTITUCIONAL USS
# ==============================================================================
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import ipywidgets as widgets
from ipywidgets import interact

USS_BLUE = '#00205B'          # Primario institucional
USS_GOLD = '#D4AF37'          # Secundario dorado
USS_ACCENT_BLUE = '#1E88E5'   # Azul de realce
USS_ACCENT_GREEN = '#27AE60'  # Verde de realce
USS_ACCENT_RED = '#C0392B'    # Rojo de realce (vector normal)
USS_GRAY = '#7F8C8D'          # Gris auxiliar
USS_LIGHT_BG = '#F8F9FA'      # Fondo claro

print("✅ Entorno configurado con Poly3DCollection y Paleta USS.")
```

```python
# ==============================================================================
# 📐 DEMOSTRACIÓN SIMBÓLICA CON SYMPY: LAGRANGE Y ORTOGONALIDAD
# ==============================================================================
u1, u2, u3 = sp.symbols('u_1 u_2 u_3', real=True)
v1, v2, v3 = sp.symbols('v_1 v_2 v_3', real=True)
w1, w2, w3 = sp.symbols('w_1 w_2 w_3', real=True)

U = sp.Matrix([u1, u2, u3])
V = sp.Matrix([v1, v2, v3])
W = sp.Matrix([w1, w2, w3])

# Producto cruz simbólico mediante fórmula determinante
cross_simb = sp.Matrix([
    u2*v3 - u3*v2,
    u3*v1 - u1*v3,
    u1*v2 - u2*v1
])

# 1. Ortogonalidad con U y V
dot_u = sp.simplify(cross_simb.dot(U))
dot_v = sp.simplify(cross_simb.dot(V))
print(f"Demostración analítica (u x v) · u = {dot_u}")
print(f"Demostración analítica (u x v) · v = {dot_v}")
assert dot_u == 0 and dot_v == 0

# 2. Identidad de Lagrange: ||u x v||² - (||u||² ||v||² - (u·v)²) == 0
norm_cross_sq = cross_simb.dot(cross_simb)
lagrange_rhs = (U.dot(U)) * (V.dot(V)) - (U.dot(V))**2
diff_lagrange = sp.simplify(norm_cross_sq - lagrange_rhs)
print(f"Verificación Identidad de Lagrange: diferencia = {diff_lagrange}")
assert diff_lagrange == 0

# 3. Triple producto escalar simbólico
det_mixto = sp.Matrix([
    [u1, u2, u3],
    [v1, v2, v3],
    [w1, w2, w3]
]).det()
print("Triple producto escalar [u, v, w] =")
sp.pprint(det_mixto)
print("✅ Todas las propiedades teóricas demostradas formalmente.")
```

```python
# ==============================================================================
# 🎮 SIMULADOR 3D INTERACTIVO: PARALELEPÍPEDO CON POLY3DCOLLECTION
# ==============================================================================

def simular_paralelepipedo_3d(ux=3.0, uy=0.0, uz=1.0,
                               vx=1.0, vy=2.0, vz=0.0,
                               wx=0.0, wy=1.0, wz=4.0):
    u = np.array([ux, uy, uz], dtype=float)
    v = np.array([vx, vy, vz], dtype=float)
    w = np.array([wx, wy, wz], dtype=float)
    
    # Cálculos
    uxv = np.cross(u, v)
    area_paral = float(np.linalg.norm(uxv))
    area_tri = 0.5 * area_paral
    
    mat_mixto = np.array([u, v, w])
    det_val = float(np.linalg.det(mat_mixto))
    vol_paral = abs(det_val)
    vol_tet = vol_paral / 6.0
    
    # 8 Vértices del paralelepípedo
    v0 = np.array([0., 0., 0.])
    v1 = u
    v2 = u + v
    v3 = v
    v4 = w
    v5 = u + w
    v6 = u + v + w
    v7 = v + w
    
    caras = [
        [v0, v1, v2, v3],  # Base inferior
        [v4, v5, v6, v7],  # Base superior
        [v0, v1, v5, v4],  # Cara frontal
        [v2, v3, v7, v6],  # Cara trasera
        [v0, v3, v7, v4],  # Cara lateral izq
        [v1, v2, v6, v5]   # Cara lateral der
    ]
    
    fig = plt.figure(figsize=(14, 6), facecolor='white')
    
    # Subplot 3D
    ax = fig.add_subplot(1, 2, 1, projection='3d')
    ax.set_title("Paralelepípedo 3D Semitransparente", fontsize=11, fontweight='bold', color=USS_BLUE)
    
    # Colección de caras con Poly3DCollection
    polys = Poly3DCollection(caras, alpha=0.25, edgecolor=USS_BLUE, linewidths=1.2)
    polys.set_facecolor(USS_GOLD)
    ax.add_collection3d(polys)
    
    # Base u-v resaltada
    base_poly = Poly3DCollection([caras[0]], alpha=0.45, edgecolor=USS_BLUE, linewidths=2.0)
    base_poly.set_facecolor(USS_ACCENT_BLUE)
    ax.add_collection3d(base_poly)
    
    # Vectores Arista
    ax.quiver(0, 0, 0, u[0], u[1], u[2], color=USS_BLUE, linewidth=3.5, arrow_length_ratio=0.12, label="u (base 1)")
    ax.quiver(0, 0, 0, v[0], v[1], v[2], color=USS_GOLD, linewidth=3.5, arrow_length_ratio=0.12, label="v (base 2)")
    ax.quiver(0, 0, 0, w[0], w[1], w[2], color=USS_ACCENT_GREEN, linewidth=3.5, arrow_length_ratio=0.12, label="w (altura)")
    
    # Vector normal u x v
    escala = min(1.0, max(np.linalg.norm(u), np.linalg.norm(v), np.linalg.norm(w)) / (area_paral + 1e-9))
    uxv_vis = uxv * escala
    ax.quiver(0, 0, 0, uxv_vis[0], uxv_vis[1], uxv_vis[2], color=USS_ACCENT_RED, linewidth=3.0, arrow_length_ratio=0.15, label="u x v (normal)")
    
    all_pts = np.array([v0, v1, v2, v3, v4, v5, v6, v7, uxv_vis])
    margin = 0.5
    ax.set_xlim(all_pts[:, 0].min() - margin, all_pts[:, 0].max() + margin)
    ax.set_ylim(all_pts[:, 1].min() - margin, all_pts[:, 1].max() + margin)
    ax.set_zlim(all_pts[:, 2].min() - margin, all_pts[:, 2].max() + margin)
    ax.set_xlabel('X', fontweight='bold', color=USS_BLUE)
    ax.set_ylabel('Y', fontweight='bold', color=USS_BLUE)
    ax.set_zlabel('Z', fontweight='bold', color=USS_BLUE)
    ax.legend(loc='upper left', fontsize=8)
    ax.view_init(elev=20, azim=55)
    
    # Subplot Métricas
    ax2 = fig.add_subplot(1, 2, 2)
    ax2.set_facecolor(USS_LIGHT_BG)
    ax2.set_title("Métricas Analíticas y Volúmenes", fontsize=11, fontweight='bold', color=USS_BLUE)
    
    texto = (
        "📐 DATOS DE ENTRADA:\n"
        f" • u = ({ux:g}, {uy:g}, {uz:g})\n"
        f" • v = ({vx:g}, {vy:g}, {vz:g})\n"
        f" • w = ({wx:g}, {wy:g}, {wz:g})\n\n"
        "✖️ PRODUCTO CRUZ u x v:\n"
        f" • u x v = ({uxv[0]:.2f}, {uxv[1]:.2f}, {uxv[2]:.2f})\n"
        f" • ||u x v|| = {area_paral:.4f}\n"
        f" • (u x v) · u = {np.dot(uxv, u):.2e} (Ortogonal)\n"
        f" • (u x v) · v = {np.dot(uxv, v):.2e} (Ortogonal)\n\n"
        "📊 ÁREAS Y VOLÚMENES:\n"
        f" • Área Paralelogramo Base = {area_paral:.4f} [u²]\n"
        f" • Área Triángulo Sustentado = {area_tri:.4f} [u²]\n"
        f" • Triple Producto [u, v, w] = {det_val:.4f}\n"
        f" • Volumen Paralelepípedo V  = {vol_paral:.4f} [u³]\n"
        f" • Volumen Tetraedro V_tet   = {vol_tet:.4f} [u³]"
    )
    ax2.text(0.08, 0.92, texto, transform=ax2.transAxes, fontsize=10.5, verticalalignment='top',
             fontfamily='monospace', bbox=dict(boxstyle='round,pad=0.8', facecolor='white', edgecolor=USS_GOLD, linewidth=2.0))
    ax2.axis('off')
    
    plt.tight_layout()
    plt.show()

interact(simular_paralelepipedo_3d,
         ux=widgets.FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description='u_x:'),
         uy=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='u_y:'),
         uz=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='u_z:'),
         vx=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='v_x:'),
         vy=widgets.FloatSlider(value=2.0, min=-5.0, max=5.0, step=0.5, description='v_y:'),
         vz=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='v_z:'),
         wx=widgets.FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='w_x:'),
         wy=widgets.FloatSlider(value=1.0, min=-5.0, max=5.0, step=0.5, description='w_y:'),
         wz=widgets.FloatSlider(value=4.0, min=-5.0, max=5.0, step=0.5, description='w_z:'));
```

> En mecánica clásica y robótica:
> - El **torque** respecto a un punto $O$ generado por una fuerza $\mathbf{F}$ aplicada en una posición $\mathbf{r}$ se define como el producto cruz:
>   $$\boldsymbol{\tau} = \mathbf{r} \times \mathbf{F}$$
> - El paralelepípedo sustentado por tres fuerzas o velocidades angulares concurrentes modela el espacio accesible y la rigidez torsional del efector final de un brazo robótico.

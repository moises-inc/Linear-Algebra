## 1. Fundamentos Teóricos

### 1.1. Vector y Norma Euclidiana
Un vector en $\mathbb{R}^n$ (=2,3$) se representa algebraicamente como una hBctupla ordenada de componentes reales:
17830\mathbf{u} = (u_1, u_2, \dots, u_n) \in \mathbb{R}^n17830
Su **norma euclidiana** (longitud o magnitud geométrica) se define formalmente como:
17830\|\mathbf{u}\| = \sqrt{\sum_{i=1}^n u_i^2}17830

El **vector unitario** (versor o dirección normalizada) para cualquier vector no nulo $\mathbf{u} \neq \mathbf{0}$ se calcula mediante:
17830\hat{\mathbf{u}} = \frac{\mathbf{u}}{\|\mathbf{u}\|}, \quad \text{con } \|\hat{\mathbf{u}}\| = 117830

---

### 1.2. Producto Escalar (Punto) y Ángulo Subtendido
El producto escalar euclidiano combina el álgebra de componentes y la métrica angular:
17830\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^n u_i v_i = \|\mathbf{u}\| \|\mathbf{v}\| \cos\theta17830

Despejando el ángulo no dirigido $\theta \in [0, \pi]$:
17830\theta = \arccos\left( \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|} \right)17830

> [!IMPORTANT] Teoremas Fundamentales del Producto Escalar
> - **Condición de Ortogonalidad:** Dos vectores no nulos son perpendiculares si y solo si su producto escalar es nulo:
>   17830\mathbf{u} \perp \mathbf{v} \iff \mathbf{u} \cdot \mathbf{v} = 017830
> - **Desigualdad de Cauchy-Schwarz:** Para todo par de vectores $\mathbf{u}, \mathbf{v} \in \mathbb{R}^n$, se cumple:
>   17830|\mathbf{u} \cdot \mathbf{v}| \le \|\mathbf{u}\| \|\mathbf{v}\|17830
>   La igualdad estricta se alcanza si y solo si los vectores son linealmente dependientes (colineales).

---

### 1.3. Descomposición y Proyección Ortogonal
Dado un vector $\mathbf{u}$ y una dirección no nula $\mathbf{v} \neq \mathbf{0}$, la **proyección ortogonal** de $\mathbf{u}$ sobre $\mathbf{v}$ es la sombra proyectada a lo largo del subespacio generado por $\mathbf{v}$:
17830\operatorname{proj}_{\mathbf{v}}(\mathbf{u}) = \left( \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{v}\|^2} \right) \mathbf{v}17830

La **componente ortogonal** (vector residual perpendicular a $\mathbf{v}$) se define como:
17830\mathbf{u}_\perp = \mathbf{u} - \operatorname{proj}_{\mathbf{v}}(\mathbf{u})17830

> [!NOTE] Propiedades Invariantes de la Descomposición
> 1. **Reconstrucción Exacta:** $\mathbf{u} = \operatorname{proj}_{\mathbf{v}}(\mathbf{u}) + \mathbf{u}_\perp$.
> 2. **Ortogonalidad Estricta:** $\mathbf{u}_\perp \cdot \mathbf{v} = 0$.
> 3. **Teorema de Pitágoras Vectorial:** $\|\mathbf{u}\|^2 = \|\operatorname{proj}_{\mathbf{v}}(\mathbf{u})\|^2 + \|\mathbf{u}_\perp\|^2$.

---

### 1.4. Cosenos Directores en $\mathbb{R}^3$
Para $\mathbf{v} = (v_1, v_2, v_3) \in \mathbb{R}^3$, los ángulos directores $\alpha, \beta, \gamma$ que forma con los ejes canónicos , Y, Z$ satisfacen:
17830\cos\alpha = \frac{v_1}{\|\mathbf{v}\|}, \quad \cos\beta = \frac{v_2}{\|\mathbf{v}\|}, \quad \cos\gamma = \frac{v_3}{\|\mathbf{v}\|}17830
cumpliéndose la identidad pitagórica fundamental:
17830\cos^2\alpha + \cos^2\beta + \cos^2\gamma = 117830

## 2. Implementación Computacional

En esta sección se implementan los algoritmos analíticos y se realiza la demostración formal simbólica con SymPy, seguida de las verificaciones numéricas y generación de figuras individuales en 2D y 3D.


```
def norma_vector(v: np.ndarray) -> float:
    """Calcula la norma euclidiana del vector v."""
    return float(np.linalg.norm(v))

def vector_unitario(v: np.ndarray) -> np.ndarray:
    """Calcula el versor unitario."""
    n = norma_vector(v)
    if np.isclose(n, 0.0):
        raise ValueError("No se puede normalizar el vector nulo.")
    return v / n

def producto_punto(u: np.ndarray, v: np.ndarray) -> float:
    """Calcula el producto escalar euclidiano."""
    return float(np.dot(u, v))

def angulo_entre_vectores(u: np.ndarray, v: np.ndarray) -> tuple[float, float]:
    """Calcula el angulo no dirigido en radianes y grados."""
    nu, nv = norma_vector(u), norma_vector(v)
    cos_theta = np.clip(producto_punto(u, v) / (nu * nv), -1.0, 1.0)
    theta_rad = float(np.arccos(cos_theta))
    return theta_rad, float(np.degrees(theta_rad))

def proyeccion_ortogonal(u: np.ndarray, v: np.ndarray) -> np.ndarray:
    """Calcula la proyeccion ortogonal de u sobre v."""
    nv2 = producto_punto(v, v)
    return (producto_punto(u, v) / nv2) * v

def componente_ortogonal(u: np.ndarray, v: np.ndarray) -> np.ndarray:
    """Calcula la componente ortogonal u_perp = u - proj_v(u)."""
    return u - proyeccion_ortogonal(u, v)

def cosenos_directores_3d(v: np.ndarray) -> dict:
    """Calcula cosenos y angulos directores en R3."""
    nv = norma_vector(v)
    cos_a, cos_b, cos_c = v[0] / nv, v[1] / nv, v[2] / nv
    return {
        'cos_alpha': float(cos_a), 'cos_beta': float(cos_b), 'cos_gamma': float(cos_c),
        'alpha_deg': float(np.degrees(np.arccos(np.clip(cos_a, -1.0, 1.0)))),
        'beta_deg': float(np.degrees(np.arccos(np.clip(cos_b, -1.0, 1.0)))),
        'gamma_deg': float(np.degrees(np.arccos(np.clip(cos_c, -1.0, 1.0)))),
        'suma_cuadrados': float(cos_a**2 + cos_b**2 + cos_c**2)
    }
```

```
# Demostracion simbolica con SymPy
u1, u2, u3 = sp.symbols('u1 u2 u3', real=True)
v1, v2, v3 = sp.symbols('v1 v2 v3', real=True)

u_sym = sp.Matrix([u1, u2, u3])
v_sym = sp.Matrix([v1, v2, v3])

dot_sym = u_sym.dot(v_sym)
norm_v_sq = v_sym.dot(v_sym)
proj_sym = (dot_sym / norm_v_sq) * v_sym
perp_sym = u_sym - proj_sym

# Verificacion analitica de ortogonalidad
ortog_sym = sp.simplify(perp_sym.dot(v_sym))
print("Formula simbolica de proj_v(u):")
display(proj_sym)
print(f"Verificacion de ortogonalidad analitica (u_perp . v): {ortog_sym}")
assert ortog_sym == 0
```

```
# Evaluacion numerica en R2 y R3
u_2d = np.array([4.0, 3.0])
v_2d = np.array([6.0, 1.0])

proj_2d = proyeccion_ortogonal(u_2d, v_2d)
perp_2d = componente_ortogonal(u_2d, v_2d)
theta_rad, theta_deg = angulo_entre_vectores(u_2d, v_2d)

print("--- Caso Numerico en R2 ---")
print(f"Vector u: {u_2d}")
print(f"Vector v: {v_2d}")
print(f"Norma ||u||: {norma_vector(u_2d):.4f}, Norma ||v||: {norma_vector(v_2d):.4f}")
print(f"Producto punto u . v: {producto_punto(u_2d, v_2d):.4f}")
print(f"Angulo subtendido: {theta_deg:.2f} deg ({theta_rad:.4f} rad)")
print(f"Proyeccion proj_v(u): {proj_2d}")
print(f"Componente u_perp:    {perp_2d}")
print(f"Verificacion u_perp . v: {producto_punto(perp_2d, v_2d):.2e}")

# Pitagoras
norm_u_sq = norma_vector(u_2d)**2
norm_decomp = norma_vector(proj_2d)**2 + norma_vector(perp_2d)**2
print(f"Pitagoras: ||u||^2 = {norm_u_sq:.4f}, ||proj||^2 + ||u_perp||^2 = {norm_decomp:.4f}")
```

## 3. Discusión Geométrica de Resultados

1. **Descomposición Canónica:** Todo vector $\mathbf{u}$ en un espacio con producto interno euclidiano admite una descomposición única en dos vectores ortogonales entre sí: uno perteneciente a la recta generada por $\mathbf{v}$ ($\operatorname{proj}_{\mathbf{v}}(\mathbf{u})$) y otro en su complemento ortogonal ($\mathbf{u}_\perp$).
2. **Propiedad de Mínima Distancia:** El vector proyección $\operatorname{proj}_{\mathbf{v}}(\mathbf{u})$ representa el punto sobre la recta soporte de $\mathbf{v}$ que minimiza la distancia euclidiana hacia $\mathbf{u}$. En consecuencia, la norma $\|\mathbf{u}_\perp\|$ mide la distancia perpendicular mínima entre $\mathbf{u}$ y la recta soporte.
3. **Invarianza Pitagórica:** La identidad $\|\mathbf{u}\|^2 = \|\operatorname{proj}_{\mathbf{v}}(\mathbf{u})\|^2 + \|\mathbf{u}_\perp\|^2$ se verifica numéricamente con un error residual acotado por 0^{-15}$, demostrando la estabilidad numérica del método algebraico implementado.


```
# Explorador Interactivo de Proyecciones 2D
def actualizar_proyeccion(ux=4.0, uy=3.0, vx=6.0, vy=1.0):
    u = np.array([ux, uy], dtype=float)
    v = np.array([vx, vy], dtype=float)
    if np.isclose(np.linalg.norm(v), 0.0):
        print("El vector v no puede ser nulo.")
        return
    proj = proyeccion_ortogonal(u, v)
    perp = componente_ortogonal(u, v)
    _, theta_deg = angulo_entre_vectores(u, v)
    
    fig, ax = plt.subplots(figsize=(7, 6), facecolor='white')
    ax.set_facecolor(USS_LIGHT_BG)
    ax.axhline(0, color=USS_GRAY, linewidth=0.8, linestyle='--')
    ax.axvline(0, color=USS_GRAY, linewidth=0.8, linestyle='--')
    
    # Recta soporte
    v_dir = vector_unitario(v)
    t_vals = np.linspace(-3, np.linalg.norm(v) + 3, 50)
    ax.plot(t_vals * v_dir[0], t_vals * v_dir[1], color=USS_GRAY, linestyle=':', linewidth=1.0)
    
    # Vectores
    ax.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color=USS_BLUE, width=0.016, label="u")
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=USS_GOLD, width=0.016, label="v")
    ax.quiver(0, 0, proj[0], proj[1], angles='xy', scale_units='xy', scale=1, color=USS_ACCENT_GREEN, width=0.018, label="proj_v(u)")
    ax.plot([u[0], proj[0]], [u[1], proj[1]], color=USS_ACCENT_RED, linestyle='--', linewidth=1.8, label="u_perp")
    
    ax.set_title(f"Proyeccion Dinamica: theta = {theta_deg:.1f} deg | ||proj|| = {norma_vector(proj):.2f}",
                 fontsize=11, fontweight='bold', color=USS_BLUE)
    ax.set_aspect('equal')
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='upper left', fontsize=8.5)
    plt.tight_layout()
    plt.show()

interact(actualizar_proyeccion,
         ux=widgets.FloatSlider(value=4.0, min=-8.0, max=8.0, step=0.5, description='u_x:'),
         uy=widgets.FloatSlider(value=3.0, min=-8.0, max=8.0, step=0.5, description='u_y:'),
         vx=widgets.FloatSlider(value=6.0, min=-8.0, max=8.0, step=0.5, description='v_x:'),
         vy=widgets.FloatSlider(value=1.0, min=-8.0, max=8.0, step=0.5, description='v_y:'));
```

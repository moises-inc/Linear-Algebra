## 📚 1. Fundamentos Teóricos y Matemáticos

### 1.1. Vector y Norma Euclidiana
Un vector en $\mathbb{R}^n$ ($n=2,3$) se representa algebraicamente como una $n$-tupla de escalares reales:
$$\mathbf{u} = (u_1, u_2, \dots, u_n) \in \mathbb{R}^n$$
Su **norma euclidiana** (magnitud o longitud) se define mediante la raíz de la suma de sus componentes al cuadrado:
$$\|\mathbf{u}\| = \sqrt{\sum_{i=1}^n u_i^2}$$

El **vector unitario** (versor o normalización) en la dirección de $\mathbf{u} \neq \mathbf{0}$ es:
$$\hat{\mathbf{u}} = \frac{\mathbf{u}}{\|\mathbf{u}\|}, \quad \text{con } \|\hat{\mathbf{u}}\| = 1$$

---

### 1.2. Producto Escalar (Punto) y Ángulo $\theta$
El producto punto entre $\mathbf{u}$ y $\mathbf{v}$ combina álgebra y trigonometría:
$$\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^n u_i v_i = \|\mathbf{u}\| \|\mathbf{v}\| \cos\theta$$

Por consiguiente, el ángulo no dirigido $\theta \in [0, \pi]$ se obtiene despejando:
$$\theta = \arccos\left( \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|} \right)$$

> [!IMPORTANT] Criterio de Ortogonalidad y Paralelismo
> - **Ortogonalidad:** Dos vectores no nulos son perpendiculares si y solo si $\mathbf{u} \cdot \mathbf{v} = 0$ (ya que $\cos(90^\circ) = 0$).
> - **Paralelismo:** Dos vectores son paralelos si $\mathbf{u} = c\,\mathbf{v}$ ($c \neq 0$), lo que equivale a $|\mathbf{u} \cdot \mathbf{v}| = \|\mathbf{u}\| \|\mathbf{v}\|$ ($\theta = 0^\circ$ o $180^\circ$).
> - **Desigualdad de Cauchy-Schwarz:** Para todo par de vectores $\mathbf{u}, \mathbf{v} \in \mathbb{R}^n$, se cumple $|\mathbf{u} \cdot \mathbf{v}| \le \|\mathbf{u}\| \|\mathbf{v}\|$.

---

### 1.3. Proyección Ortogonal y Descomposición
Dado un vector $\mathbf{u}$ y un vector director de referencia $\mathbf{v} \neq \mathbf{0}$, la **proyección ortogonal** de $\mathbf{u}$ sobre $\mathbf{v}$ representa la sombra geométrica de $\mathbf{u}$ a lo largo de la recta soporte de $\mathbf{v}$:
$$\operatorname{proj}_{\mathbf{v}}(\mathbf{u}) = \left( \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{v}\|^2} \right) \mathbf{v}$$

La **componente ortogonal** (o vector perpendicular residual) es:
$$\mathbf{u}_\perp = \mathbf{u} - \operatorname{proj}_{\mathbf{v}}(\mathbf{u})$$

> [!NOTE] Propiedades Invariantes de la Descomposición
> 1. **Reconstrucción Exacta:** $\mathbf{u} = \operatorname{proj}_{\mathbf{v}}(\mathbf{u}) + \mathbf{u}_\perp$.
> 2. **Ortogonalidad Estricta:** $\mathbf{u}_\perp \cdot \mathbf{v} = 0$.
> 3. **Teorema de Pitágoras Vectorial:** $\|\mathbf{u}\|^2 = \|\operatorname{proj}_{\mathbf{v}}(\mathbf{u})\|^2 + \|\mathbf{u}_\perp\|^2$.

---

### 1.4. Cosenos Directores en $\mathbb{R}^3$
Para $\mathbf{v} = (v_1, v_2, v_3) \in \mathbb{R}^3$, los ángulos $\alpha, \beta, \gamma$ que forma con los ejes canónicos $X, Y, Z$ satisfacen:
$$\cos\alpha = \frac{v_1}{\|\mathbf{v}\|}, \quad \cos\beta = \frac{v_2}{\|\mathbf{v}\|}, \quad \cos\gamma = \frac{v_3}{\|\mathbf{v}\|}$$
cumpliéndose la identidad pitagórica fundamental:
$$\cos^2\alpha + \cos^2\beta + \cos^2\gamma = 1$$


```python
# ==============================================================================
# 🛠️ CONFIGURACIÓN DEL ENTORNO Y PALETA INSTITUCIONAL USS
# ==============================================================================
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import ipywidgets as widgets
from ipywidgets import interact

# Paleta Institucional USS
USS_BLUE = '#00205B'          # Primario institucional
USS_GOLD = '#D4AF37'          # Secundario dorado
USS_ACCENT_BLUE = '#1E88E5'   # Azul de realce
USS_ACCENT_GREEN = '#27AE60'  # Verde de realce / proyección
USS_ACCENT_RED = '#C0392B'    # Rojo de alerta / componente ortogonal
USS_GRAY = '#7F8C8D'          # Gris auxiliar
USS_LIGHT_BG = '#F8F9FA'      # Fondo claro

print("✅ Entorno de computación científica listo con Paleta Institucional USS.")
```

```python
# ==============================================================================
# 📐 FUNCIONES VECTORIALES Y DEMOSTRACIÓN SIMBÓLICA CON SYMPY
# ==============================================================================

def calcular_propiedades_vectoriales(u, v):
    '''Calcula exhaustivamente todas las métricas analíticas entre u y v.'''
    u = np.array(u, dtype=float)
    v = np.array(v, dtype=float)
    
    norm_u = float(np.linalg.norm(u))
    norm_v = float(np.linalg.norm(v))
    dot_uv = float(np.dot(u, v))
    
    cos_theta = np.clip(dot_uv / (norm_u * norm_v), -1.0, 1.0)
    theta_rad = float(np.arccos(cos_theta))
    theta_deg = float(np.degrees(theta_rad))
    
    proj_v_u = (dot_uv / (norm_v**2)) * v
    u_perp = u - proj_v_u
    
    norm_proj = float(np.linalg.norm(proj_v_u))
    norm_perp = float(np.linalg.norm(u_perp))
    
    # Verificaciones
    ortog_err = abs(float(np.dot(u_perp, v)))
    pitagoras_err = abs(norm_u**2 - (norm_proj**2 + norm_perp**2))
    
    return {
        'norm_u': norm_u, 'norm_v': norm_v,
        'dot_uv': dot_uv, 'theta_deg': theta_deg, 'theta_rad': theta_rad,
        'proj': proj_v_u, 'u_perp': u_perp,
        'norm_proj': norm_proj, 'norm_perp': norm_perp,
        'ortog_err': ortog_err, 'pitagoras_err': pitagoras_err
    }

# Demostración simbólica con SymPy
print("--- DEMOSTRACIÓN SIMBÓLICA FORMAL ---")
u1, u2, u3 = sp.symbols('u_1 u_2 u_3', real=True)
v1, v2, v3 = sp.symbols('v_1 v_2 v_3', real=True)

U = sp.Matrix([u1, u2, u3])
V = sp.Matrix([v1, v2, v3])

# proj_V(U) = (U · V / ||V||²) * V
dot_UV = U.dot(V)
norm_V_sq = V.dot(V)
proj_simb = (dot_UV / norm_V_sq) * V
perp_simb = U - proj_simb

# Comprobar analíticamente que perp_simb · V es idénticamente CERO
ortogonalidad = sp.simplify(perp_simb.dot(V))
print(f"Producto escalar formal de u_perp con v: {ortogonalidad}")
assert ortogonalidad == 0
print("✅ Q.E.D. Se demuestra analíticamente que u_perp es ortogonal a v para todo vector en R³.")
```

```python
# ==============================================================================
# 🎮 SIMULADOR INTERACTIVO 2D: SUMA, RESTA Y PROYECCIÓN ORTOGONAL
# ==============================================================================

def simular_vectores_2d(ux=4.0, uy=3.0, vx=6.0, vy=1.0):
    u = np.array([ux, uy], dtype=float)
    v = np.array([vx, vy], dtype=float)
    
    if np.isclose(np.linalg.norm(v), 0.0):
        print("⚠️ El vector v no puede ser nulo para proyectar sobre él.")
        return
    
    props = calcular_propiedades_vectoriales(u, v)
    proj = props['proj']
    perp = props['u_perp']
    suma = u + v
    resta = u - v
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6), facecolor='white')
    
    # ------------------ PANEL 1: ÁLGEBRA VECTORIAL ------------------
    ax1.set_facecolor(USS_LIGHT_BG)
    ax1.set_title("1. Regla del Paralelogramo (Suma y Resta)", fontsize=11, fontweight='bold', color=USS_BLUE)
    ax1.axhline(0, color=USS_GRAY, linestyle='--', linewidth=0.8)
    ax1.axvline(0, color=USS_GRAY, linestyle='--', linewidth=0.8)
    
    # Vectores u y v
    ax1.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color=USS_BLUE, width=0.018, label=f"u = ({ux:g}, {uy:g})")
    ax1.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=USS_GOLD, width=0.018, label=f"v = ({vx:g}, {vy:g})")
    
    # Paralelogramo
    ax1.plot([u[0], suma[0]], [u[1], suma[1]], color=USS_GOLD, linestyle=':', linewidth=1.5)
    ax1.plot([v[0], suma[0]], [v[1], suma[1]], color=USS_BLUE, linestyle=':', linewidth=1.5)
    
    # Suma y Resta
    ax1.quiver(0, 0, suma[0], suma[1], angles='xy', scale_units='xy', scale=1, color=USS_ACCENT_GREEN, width=0.016, label=f"u+v = ({suma[0]:g}, {suma[1]:g})")
    ax1.quiver(v[0], v[1], resta[0], resta[1], angles='xy', scale_units='xy', scale=1, color=USS_ACCENT_RED, width=0.014, label="u-v")
    
    lims = [0, u[0], v[0], suma[0], resta[0]]
    lims_y = [0, u[1], v[1], suma[1], resta[1]]
    m = 2.0
    ax1.set_xlim(min(lims) - m, max(lims) + m)
    ax1.set_ylim(min(lims_y) - m, max(lims_y) + m)
    ax1.set_aspect('equal')
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.legend(loc='upper left', fontsize=8)
    
    # ------------------ PANEL 2: PROYECCIÓN ORTOGONAL ------------------
    ax2.set_facecolor(USS_LIGHT_BG)
    ax2.set_title("2. Descomposición Ortogonal u = proj_v(u) + u_perp", fontsize=11, fontweight='bold', color=USS_BLUE)
    ax2.axhline(0, color=USS_GRAY, linestyle='--', linewidth=0.8)
    ax2.axvline(0, color=USS_GRAY, linestyle='--', linewidth=0.8)
    
    # Línea soporte
    v_norm = v / np.linalg.norm(v)
    t_vals = np.linspace(-3, max(props['norm_v'], props['norm_proj']) + 3, 50)
    ax2.plot(t_vals * v_norm[0], t_vals * v_norm[1], color=USS_GRAY, linestyle='-.', linewidth=1.0, label="Soporte de v")
    
    ax2.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color=USS_BLUE, width=0.018, label="u")
    ax2.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color=USS_GOLD, width=0.018, label="v")
    ax2.quiver(0, 0, proj[0], proj[1], angles='xy', scale_units='xy', scale=1, color=USS_ACCENT_GREEN, width=0.022, label=f"proj_v(u) = ({proj[0]:.2f}, {proj[1]:.2f})")
    
    # Caída ortogonal y componente perpendicular
    ax2.plot([u[0], proj[0]], [u[1], proj[1]], color=USS_ACCENT_RED, linestyle=':', linewidth=2.0, label="Caída ortogonal")
    ax2.quiver(0, 0, perp[0], perp[1], angles='xy', scale_units='xy', scale=1, color=USS_ACCENT_RED, width=0.016, label=f"u_perp = ({perp[0]:.2f}, {perp[1]:.2f})")
    
    all_x = [0, u[0], v[0], proj[0], perp[0]]
    all_y = [0, u[1], v[1], proj[1], perp[1]]
    ax2.set_xlim(min(all_x) - 2, max(all_x) + 2)
    ax2.set_ylim(min(all_y) - 2, max(all_y) + 2)
    ax2.set_aspect('equal')
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='upper left', fontsize=8)
    
    plt.tight_layout()
    plt.show()
    
    print(f"📊 MÉTRICAS ANALÍTICAS:")
    print(f" • ||u|| = {props['norm_u']:.4f},  ||v|| = {props['norm_v']:.4f}")
    print(f" • u · v = {props['dot_uv']:.4f}")
    print(f" • Ángulo theta = {props['theta_deg']:.2f}° ({props['theta_rad']:.4f} rad)")
    print(f" • ||proj_v(u)|| = {props['norm_proj']:.4f},  ||u_perp|| = {props['norm_perp']:.4f}")
    print(f" • Ortogonalidad u_perp · v = {props['ortog_err']:.2e} (Debe ser ~0)")
    print(f" • Pitágoras: ||u||² = {props['norm_u']**2:.4f}, ||proj||² + ||u_perp||² = {props['norm_proj']**2 + props['norm_perp']**2:.4f}")

# Controles interactivos ipywidgets
interact(simular_vectores_2d,
         ux=widgets.FloatSlider(value=4.0, min=-8.0, max=8.0, step=0.5, description='u_x:'),
         uy=widgets.FloatSlider(value=3.0, min=-8.0, max=8.0, step=0.5, description='u_y:'),
         vx=widgets.FloatSlider(value=6.0, min=-8.0, max=8.0, step=0.5, description='v_x:'),
         vy=widgets.FloatSlider(value=1.0, min=-8.0, max=8.0, step=0.5, description='v_y:'));
```

```python
# ==============================================================================
# 🌐 SIMULADOR 3D: PROYECCIONES EN R³ Y COSENOS DIRECTORES
# ==============================================================================

def simular_vectores_3d(ux=3.0, uy=-2.0, uz=5.0, vx=1.0, vy=4.0, vz=2.0):
    u = np.array([ux, uy, uz], dtype=float)
    v = np.array([vx, vy, vz], dtype=float)
    
    if np.isclose(np.linalg.norm(v), 0.0):
        print("⚠️ El vector v no puede ser el vector nulo.")
        return
        
    props = calcular_propiedades_vectoriales(u, v)
    proj = props['proj']
    perp = props['u_perp']
    
    fig = plt.figure(figsize=(12, 6), facecolor='white')
    ax = fig.add_subplot(1, 1, 1, projection='3d')
    ax.set_title("Proyección Ortogonal en R³ y Cosenos Directores", fontsize=12, fontweight='bold', color=USS_BLUE)
    
    # Vectores Quiver 3D
    ax.quiver(0, 0, 0, u[0], u[1], u[2], color=USS_BLUE, linewidth=3.0, arrow_length_ratio=0.1, label=f"u = ({ux:g}, {uy:g}, {uz:g})")
    ax.quiver(0, 0, 0, v[0], v[1], v[2], color=USS_GOLD, linewidth=3.0, arrow_length_ratio=0.1, label=f"v = ({vx:g}, {vy:g}, {vz:g})")
    ax.quiver(0, 0, 0, proj[0], proj[1], proj[2], color=USS_ACCENT_GREEN, linewidth=3.5, arrow_length_ratio=0.1, label="proj_v(u)")
    ax.quiver(proj[0], proj[1], proj[2], perp[0], perp[1], perp[2], color=USS_ACCENT_RED, linewidth=2.0, arrow_length_ratio=0.1, label="u_perp")
    
    # Línea soporte
    v_unit = v / np.linalg.norm(v)
    t = np.linspace(-2, props['norm_v'] + 2, 30)
    ax.plot(t * v_unit[0], t * v_unit[1], t * v_unit[2], color=USS_GRAY, linestyle='--', linewidth=0.8)
    
    ax.set_xlabel('X', fontweight='bold', color=USS_BLUE)
    ax.set_ylabel('Y', fontweight='bold', color=USS_BLUE)
    ax.set_zlabel('Z', fontweight='bold', color=USS_BLUE)
    ax.legend(loc='upper left', fontsize=8)
    ax.view_init(elev=25, azim=45)
    plt.tight_layout()
    plt.show()
    
    # Cosenos directores de u
    nu = props['norm_u']
    cos_a, cos_b, cos_c = u[0]/nu, u[1]/nu, u[2]/nu
    print(f"🧭 COSENOS DIRECTORES DE u:")
    print(f" • cos(alpha) = {cos_a:.4f} => alpha = {np.degrees(np.arccos(cos_a)):.2f}°")
    print(f" • cos(beta)  = {cos_b:.4f} => beta  = {np.degrees(np.arccos(cos_b)):.2f}°")
    print(f" • cos(gamma) = {cos_c:.4f} => gamma = {np.degrees(np.arccos(cos_c)):.2f}°")
    print(f" • Identidad cos²(alpha) + cos²(beta) + cos²(gamma) = {cos_a**2 + cos_b**2 + cos_c**2:.6f} (Exactamente 1.0)")

interact(simular_vectores_3d,
         ux=widgets.FloatSlider(value=3.0, min=-6.0, max=6.0, step=0.5, description='u_x:'),
         uy=widgets.FloatSlider(value=-2.0, min=-6.0, max=6.0, step=0.5, description='u_y:'),
         uz=widgets.FloatSlider(value=5.0, min=-6.0, max=6.0, step=0.5, description='u_z:'),
         vx=widgets.FloatSlider(value=1.0, min=-6.0, max=6.0, step=0.5, description='v_x:'),
         vy=widgets.FloatSlider(value=4.0, min=-6.0, max=6.0, step=0.5, description='v_y:'),
         vz=widgets.FloatSlider(value=2.0, min=-6.0, max=6.0, step=0.5, description='v_z:'));
```
